###  Poisson Models Over Specified Time Frames

#### Problem of Probability of specific arrival types

Suppose three independent Poisson processes, $N_A(t)$, $N_B(t)$, and $N_C(t)$, represent the arrivals of items of type A, B, and C, respectively. Their corresponding arrival rates are $\lambda_A$, $\lambda_B$, and $\lambda_C$ items per minute. The merged process of all arrivals is also a Poisson process with rate $\lambda = \lambda_A + \lambda_B + \lambda_C$.

When an item arrives in the merged process, the probability that it is of a specific type (e.g., type A) is given by the ratio of its individual rate to the total rate: $P(\text{Type A}) = \frac{\lambda_A}{\lambda_A + \lambda_B + \lambda_C}$. This is also true for types B and C.

Given $\lambda_A = 2.0$, $\lambda_B = 3.0$, and $\lambda_C = 5.0$ items per minute, what is the probability that the first arriving item is of type A, and exactly one of the next five arriving items is also of type A?

#### Erlang Distribution for Total Time Between Discharges

For a system with a combined Poisson process (e.g., merged processes from types A, B, and C) having a total rate of $\lambda = \lambda_A + \lambda_B + \lambda_C$, if we are interested in the time until a certain number of events occur, this often follows an Erlang distribution.

Consider a scenario where events (discharges) occur according to a Poisson process with a total rate of $\lambda = a+b+c$ items per minute. If we are interested in the total time $T$ until the 10th discharge, then $T$ follows an Erlang random variable with shape parameter $k=10$ and rate parameter $\lambda = a+b+c$.

The Probability Density Function (PDF) of $T$ is given by:
$f_T(t) = \frac{\lambda^k t^{k-1} e^{-\lambda t}}{(k-1)!} = \frac{(a+b+c)^{10} t^9 e^{-(a+b+c)t}}{9!}$ for $t \ge 0$; and $0$ otherwise.

The Expected value of $T$ is $E[T] = \frac{k}{\lambda} = \frac{10}{a+b+c}$.

The Variance of $T$ is $Var(T) = \frac{k}{\lambda^2} = \frac{10}{(a+b+c)^2}$.

#### Pictorial View: Overlapping Time Intervals in a Poisson Process

Consider a Poisson process with rate $\lambda$. We are interested in the number of arrivals within two overlapping time intervals: Interval 1: $[0, t]$, containing $N$ arrivals, & Interval 2: $[0, t+s]$, containing $M$ arrivals.

Here, the interval $[0, t]$ is entirely contained within $[0, t+s]$.

```
Time Axis: 0-------------------t-------------------t+s------------------->
           <---------- N arrivals --------->
           <------------------- M arrivals -------------------->

Visualization of Overlapping Intervals:

  [0, t]  : Events in the first interval
  |-------|
  *   *   *   *

  [0, t+s]: Events in the second (larger) interval
  |-------------------|
  *   *   *   *   *   *   *   *

The relationship: M includes all N arrivals, plus any additional arrivals in the interval (t, t+s].
```

Let $N_{[0,t]}$ be the number of arrivals in $[0, t]$ and $N_{(t, t+s]}$ be the number of arrivals in $(t, t+s]$.
Then, we can write $N = N_{[0,t]}$ and $M = N_{[0,t]} + N_{(t, t+s]}$.

Since Poisson processes have independent increments over disjoint intervals: $N_{[0,t]}$ is Poisson distributed with parameter $\lambda t$. $N_{(t, t+s]}$ is Poisson distributed with parameter $\lambda s$. $N_{[0,t]}$ and $N_{(t, t+s]}$ are independent random variables.

#### Conditional PMF of $P(M=m | N=n)$ for Overlapping Intervals

We know that $M = N + N_{\text{additional}}$, where $N_{\text{additional}}$ is the number of arrivals in the interval $(t, t+s]$. This $N_{\text{additional}}$ is a Poisson random variable with parameter $\lambda s$, and it is independent of $N$.

Given that $N=n$, we substitute this into the equation for $M$: $M = n + N_{\text{additional}}$. The event $M=m$ given $N=n$ implies: $n + N_{\text{additional}} = m$ or $N_{\text{additional}} = m - n$.

For this to be a valid number of arrivals, we must have $m \ge n$. If $m < n$, the probability is 0 because the number of arrivals in a larger interval cannot be less than the number of arrivals in a smaller sub-interval.

Therefore, the conditional probability is simply the probability that $N_{\text{additional}}$ takes the value $m-n$: $P(M=m | N=n) = P(N_{\text{additional}} = m - n)$. Since $N_{\text{additional}} \sim \text{Poisson}(\lambda s)$, its PMF is: $P(N_{\text{additional}} = k) = \frac{(\lambda s)^k e^{-\lambda s}}{k!}$. Substituting $k = m-n$
- $P(M=m | N=n) = \frac{(\lambda s)^{m-n} e^{-\lambda s}}{(m-n)!}$

This holds for $m = n, n+1, n+2, \dots$. The probability is $0$ if $m < n$.

#### Equivalence Between Number of Arrivals and Arrival Times in a Poisson Process

In a Poisson process, we can describe events using two related random variables, $N(t)$ the number of arrivals that occur in the time interval $(0, t]$, and $Y_k$ the time at which the $k$-th arrival occurs.

These two ways of describing events are closely linked, and understanding their equivalence is fundamental.

Let's analyze examples:

1. Event $\{N(5) > 1\}$ is equivalent to the event $\{Y_k \le b\}$ for suitable $b$ and $k$.

- Event $\{N(5) > 1\}$ means that the number of arrivals in the interval $(0, 5]$ is greater than 1. In other words, there are at least two arrivals by time $t=5$.
- Equivalent Arrival Time Event: If there are at least two arrivals by time $t=5$, it logically follows that the time of the 2nd arrival ($Y_2$) must have occurred at or before time $t=5$.

    Therefore, the event $\{N(5) > 1\}$ is equivalent to $\{Y_2 \le 5\}$.
    Here, $k=2$ and $b=5$.

2. Event $\{2 < Y_3 \le Y_4 \le 5\}$ is equivalent to event $\{N(2) \le a, N(5) \ge b\}$ for suitable $a$ and $b$.
- Event $\{2 < Y_3 \le Y_4 \le 5\}$:** This compound event describes two conditions about arrival times:
    *   $\{2 < Y_3\}$: The 3rd arrival occurs strictly after time $t=2$. If the 3rd arrival occurs after $t=2$, it means that by time $t=2$, there could have been at most 2 arrivals.
        Thus, this part implies $\{N(2) \le 2\}$.
    *   $\{Y_4 \le 5\}$: The 4th arrival occurs at or before time $t=5$. If the 4th arrival occurs by $t=5$, it means that by time $t=5$, there must have been at least 4 arrivals. Thus, this part implies $\{N(5) \ge 4\}$.

- Equivalent Number of Arrivals Event: Combining these two conditions, the event $\{2 < Y_3 \le Y_4 \le 5\}$ is equivalent to the event $\{N(2) \le 2 \text{ and } N(5) \ge 4\}$. Here, $a=2$ and $b=4$ (for $N(2) \le a$ and $N(5) \ge b$).

These equivalences are particularly useful when performing probability calculations, as you can choose the representation that simplifies the problem.

#### Practical Examples of Poisson Models Over Specified Time Frames

Some practical applications of the Poisson model for events occurring within defined time frames:

1.  **Arrival of Passengers to the Claim Section (e.g., at an airport or luggage claim)**:
    
Imagine an airport's baggage claim area. Passengers arrive at the claim section to pick up their luggage. This can be modeled as a Poisson process if the average rate of passenger arrivals is constant over a given observation period, and individual arrivals are independent.

**Poisson Model**: Let $\lambda$ be the average rate of passenger arrivals per unit of time (e.g., per minute). If you want to model the number of passengers arriving in a 15-minute interval ($\tau = 15$ minutes), the number of arrivals $N_{15}$ would follow a Poisson distribution with parameter $\lambda\tau = 15\lambda$. The probability of exactly $k$ passengers arriving in those 15 minutes would be $P(N_{15} = k) = \frac{(15\lambda)^k e^{-15\lambda}}{k!}$.

**Use Cases**: This model helps in staffing the claim section, predicting queue lengths, and optimizing resource allocation.

2.  **Order Arrival at a Retailer from 5:00 PM to 5:30 PM**:

This is a specific, fixed time window ($\tau = 30$ minutes). During this half-hour period, a retailer receives online or in-store orders. Assuming that the orders arrive independently and at an average constant rate during this specific window (even if the rate might differ at other times of the day).

**Poisson Model**: Let $\lambda$ be the average rate of order arrivals per minute between 5:00 PM and 5:30 PM. The duration of the interval is $\tau = 30$ minutes. The number of orders received, $N_{30}$, in this interval would follow a Poisson distribution with parameter $\lambda\tau = 30\lambda$. The probability of receiving exactly $k$ orders would be $P(N_{30} = k) = \frac{(30\lambda)^k e^{-30\lambda}}{k!}$.

**Use Cases**: This can help in predicting peak order volumes, adjusting staffing for order fulfillment or customer service during specific busy periods, and managing inventory.

3.  **Order Arrival Over the Course of a Day**:

Here, the time frame is much longer ($\tau = 24$ hours). While a single, constant $\lambda$ over an entire day might be an oversimplification (e.g., order rates are usually higher during business hours and lower at night), a Poisson model can still be applied if we consider an *average* daily rate, or if we model different segments of the day with different $\lambda$ values (a non-homogeneous Poisson process).

**Poisson Model (Simplified)**: If we assume a constant average order rate $\lambda$ over the entire 24-hour period, then the total number of orders $N_{24}$ in a day would be Poisson distributed with parameter $\lambda\tau = 24\lambda$. The probability of $k$ orders in a day would be $P(N_{24} = k) = \frac{(24\lambda)^k e^{-24\lambda}}{k!}$.

**Use Cases**: This model can be useful for long-term planning, estimating daily demand, forecasting sales, and setting overall inventory levels. For more accuracy, one might use a non-homogeneous Poisson process to account for varying rates during the day.

#### The key assumptions for a Poisson process are:

**Independence**: Events (arrivals) occur independently of each other.
**Stationarity**: The average rate of events ($\lambda$$\lambda$) is constant over the specified time interval.
**Rarity**: Events are rare relative to the total number of opportunities for them to occur (in practice, this leads to the approximation of Binomial by Poisson).

### Case Study & Generalized Problem: Items in a Common Buffer

Type A, B, and C items are placed in a common buffer, each type arriving as part of an independent Poisson process with average arrival rates, respectively, of a, b, and c items per minute.

For the first four parts of this problem, assume the buffer is discharged immediately whenever it contains a total of ten items.

- (a) The probability that, of the first ten items to arrive at the buffer, only the first and one other are type A.

- (b) The probability that any particular discharge of the buffer contains five times as many type A items as type B items.

- (c) The PDF, expectation, and variance for the total time between consecutive discharges of the buffer.

- (d) The probability that exactly two of each of the three item types arrive at the buffer input during any particular five-minute interval.


Let the arrival rates be $a=2.0$, $b=3.0$, and $c=5.0$ items per minute for Type A, B, and C respectively. The total arrival rate is $\lambda_{total} = a+b+c = 2.0+3.0+5.0 = 10.0$ items per minute.

Let $P_A = \frac{a}{a+b+c}$, $P_B = \frac{b}{a+b+c}$, and $P_C = \frac{c}{a+b+c}$ be the probabilities that any randomly chosen item from the merged stream is of type A, B, or C respectively.

````python
import math

# Illustrative arrival rates (per minute)
a = 2.0 ; b = 3.0 ; c = 5.0
total_rate = a + b + c
P_A = a / total_rate ; P_B = b / total_rate ; P_C = c / total_rate
````

##### (a) The probability that, of the first ten items to arrive at the buffer, only the first and one other are type A

We consider the first 10 items as a sequence of independent trials, where each item's type is determined by the probabilities $P_A, P_B, P_C$. The condition is that the first item is Type A, and exactly one of the remaining nine items is Type A.

- Probability of the first item being Type A: $P_A$
- Probability of exactly one of the remaining 9 items being Type A: This follows a binomial distribution with $n=9$ trials, $k=1$ success, and success probability $P_A$. The probability is $\binom{9}{1} P_A^1 (1 - P_A)^8$.

Since these two events are independent within the sequence, the total probability is the product:
$P(\text{event (a)}) = P_A \times \binom{9}{1} P_A^1 (1 - P_A)^8 = \binom{9}{1} P_A^2 (1 - P_A)^8$

````python
# Calculate for part (a)
n_trials = 9 ; k_successes = 1

P_not_A = 1 - P_A
combinations_9_1 = math.comb(n_trials, k_successes)
prob_one_of_next_9_is_A = combinations_9_1 * (P_A**k_successes) * (P_not_A**(n_trials - k_successes))
prob_part_a = P_A * prob_one_of_next_9_is_A
````

##### (b) The probability that any particular discharge of the buffer contains five times as many type A items as type B items

A discharge occurs when 10 items have arrived. Let $N_A$, $N_B$, and $N_C$ be the number of items of type A, B, and C, respectively, among these 10 items. We know $N_A + N_B + N_C = 10$. We are looking for the probability that $N_A = 5 N_B$.

We need to find integer pairs $(N_A, N_B)$ that satisfy this condition and $N_A + N_B \le 10$:
1. If $N_B = 0$, then $N_A = 5 \times 0 = 0$. This implies $N_C = 10 - 0 - 0 = 10$.
2. If $N_B = 1$, then $N_A = 5 \times 1 = 5$. This implies $N_C = 10 - 5 - 1 = 4$.
(If $N_B = 2$, $N_A = 10$, then $N_A + N_B = 12 > 10$, so no more valid pairs).

We use the Multinomial distribution formula: $P(n_A, n_B, n_C) = \frac{10!}{n_A! n_B! n_C!} P_A^{n_A} P_B^{n_B} P_C^{n_C}$.

$P(\text{event (b)}) = P(N_A=0, N_B=0, N_C=10) + P(N_A=5, N_B=1, N_C=4)$

````python
# Calculate for part (b)

# Case 1: NA=0, NB=0, NC=10
prob_case1 = (math.factorial(10) / (math.factorial(0) * math.factorial(0) * math.factorial(10))) * (P_A**0) * (P_B**0) * (P_C**10)

# Case 2: NA=5, NB=1, NC=4
prob_case2 = (math.factorial(10) / (math.factorial(5) * math.factorial(1) * math.factorial(4))) * (P_A**5) * (P_B**1) * (P_C**4)

prob_part_b = prob_case1 + prob_case2
````

##### (c) The PDF, expectation, and variance for the total time between consecutive discharges of the buffer

As discussed previously (see [Erlang Distribution for Total Time Between Discharges](#scrollTo=)), the total time until the $k$-th arrival in a Poisson process follows an Erlang distribution with shape parameter $k$ and rate parameter $\lambda$. In this case, a discharge occurs after 10 items ($k=10$) from the merged Poisson process with a total rate of $\lambda_{total} = a+b+c$.

- **PDF of $T$**: $f_T(t) = \frac{\lambda_{total}^k t^{k-1} e^{-\lambda_{total} t}}{(k-1)!} = \frac{(a+b+c)^{10} t^9 e^{-(a+b+c)t}}{9!}$ for $t \ge 0$.
- **Expectation of $T$**: $E[T] = \frac{k}{\lambda_{total}} = \frac{10}{a+b+c}$.
- **Variance of $T$**: $Var(T) = \frac{k}{\lambda_{total}^2} = \frac{10}{(a+b+c)^2}$.

````python
# Calculate for part (c)
k_erlang = 10 # Number of items for discharge
lambda_erlang = total_rate # Total arrival rate
expected_time = k_erlang / lambda_erlang
variance_time = k_erlang / (lambda_erlang**2)
````

##### (d) The probability that exactly two of each of the three item types arrive at the buffer input during any particular five minute interval.

This problem was explicitly solved earlier (see [Problem: Probability of exactly two arrivals for each type in a specified time interval](#scrollTo=)).

For each independent Poisson process, the probability of exactly $k$ arrivals in a time interval $\tau$ is given by the Poisson PMF: $P(N_\tau = k) = \frac{(\lambda\tau)^k e^{-\lambda\tau}}{k!}$.

Here, $\tau = 5$ minutes and $k=2$ for each type. The overall probability is the product of the individual probabilities due to independence:

$P(\text{event (d)}) = P(N_A(5)=2) \times P(N_B(5)=2) \times P(N_C(5)=2)$

````python
time_interval_d = 5 # minutes
k_arrivals_d = 2 # exactly two arrivals for each type

# Poisson parameter for each process
param_A_d = a * time_interval_d
param_B_d = b * time_interval_d
param_C_d = c * time_interval_d

# Probability for each type
prob_A_d = (param_A_d**k_arrivals_d * np.exp(-param_A_d)) / math.factorial(k_arrivals_d)
prob_B_d = (param_B_d**k_arrivals_d * np.exp(-param_B_d)) / math.factorial(k_arrivals_d)
prob_C_d = (param_C_d**k_arrivals_d * np.exp(-param_C_d)) / math.factorial(k_arrivals_d)

# Overall probability
prob_part_d = prob_A_d * prob_B_d * prob_C_d
````

### Case Study: Streams of Poisson Process

#### Two streams of Poisson process

Given stream 1 Poisson process with arrivals at rate $\lambda_1 = 3$/hour and stay for $1/6$ of an hour. Given also stream 2 according to Poisson process with rate $\lambda_2 = 2$/hour and stay exactly $1/2$ hour.Arrivals for stream 1 and 2 are independent processes. An observer arrives at a specific time and sees arrivals/stays of stream 1 and stream 2. We compute:
1.  P({there are no stays from stream 1})
2.  P({stays from stream 1 and stream 2 = 1})
3.  P({2 arrivals/stays from stream 2 | in 10 minutes exactly 2 arrivals/stays })
4.  Confirm if the process of total arrivals is a Poisson process (Yes because independent)

````python
import numpy as np
import math

# Define parameters for Stream 1
lambda_1 = 3.0  # arrivals per hour
duration_1 = 1/6 # hours (stay duration)

# Define parameters for Stream 2
lambda_2 = 2.0  # arrivals per hour
duration_2 = 1/2 # hours (stay duration)
````

#### Three streams with transition to Stream 3

Let's extend the previous problem. We have Stream 1 (rate $\lambda_1 = 3$/hour, stay duration $d_1 = 1/6$ hour) and Stream 2 (rate $\lambda_2 = 2$/hour, stay duration $d_2 = 1/2$ hour). When a stay's duration from either Stream 1 or Stream 2 ends, the entity can decide to either end its interaction or enter a new Stream 3. **Stream 3** is also a Poisson process with its own arrival rate $\lambda_3$ and stay duration $d_3$.

Let's assume the following:
*   Stream 3 rate: $\lambda_3 = 4$/hour, stay duration: $d_3 = 1/4$ hour
*   Transition Probability - when a stay ends:
    *   If it originated from Stream 1, the probability to enter Stream 3 is $p_{\text{to_stream3_from1}} = 1/4$.
    *   If it originated from Stream 2, the probability to enter Stream 3 is $p_{\text{to_stream3_from2}} = 1/2$.
    
All processes remain independent, and the decisions to enter Stream 3 are independent for each departing stay and independent from all other Poisson processes that define this model.

````python
# Define parameters for Stream 3
lambda_3 = 4.0  # arrivals per hour
duration_3 = 1/4 # hours (stay duration)

p_to_stream3_from1 = 1/4 # Prob. to enter Stream3 upon end a stay from Stream 1
p_to_stream3_from2 = 1/2 # Prob. to enter Stream 3 upon end a stay from Stream 2
````

#### Modeling Transitions and Effective Arrival Rates into Stream 3

When a stay's duration ends in Stream 1 or Stream 2, the entities can transition into Stream 3. Since the departure processes from independent Poisson processes are also Poisson with the same rates, we can calculate the rate at which entities effectively *arrive* into Stream 3 from these transitions.

*   **Rate of transitions from Stream 1 to Stream 3 ($\lambda_{1 \to 3}$)**:
    This is the rate at which entities depart from Stream 1 multiplied by the probability of transitioning to Stream 3 from Stream 1.
    $\lambda_{1 \to 3} = \lambda_1 \cdot p_{\text{to_stream3_from1}}$

*   **Rate of transitions from Stream 2 to Stream 3 ($\lambda_{2 \to 3}$)**:
    Similarly, this is the rate at which entities depart from Stream 2 multiplied by the probability of transitioning to Stream 3 from Stream 2.
    $\lambda_{2 \to 3} = \lambda_2 \cdot p_{\text{to_stream3_from2}}$

Since these are independent Poisson streams of transitions, they can be merged. The total effective arrival rate into Stream 3 from *all* sources (its intrinsic arrivals and transitions from Stream 1 and 2) would be:

*   **Total effective arrival rate into Stream 3 ($\lambda_{\text{total_effective_3}}$)**:
    $\lambda_{\text{total_effective_3}} = \lambda_3 + \lambda_{1 \to 3} + \lambda_{2 \to 3}$

````python
# effective arrival rate from Stream 1 to Stream 3
lambda_1_to_3 = lambda_1 * p_to_stream3_from1

# effective arrival rate from Stream 2 to Stream 3
lambda_2_to_3 = lambda_2 * p_to_stream3_from2

# the total effective arrival rate into Stream 3
lambda_total_effective_3 = lambda_3 + lambda_1_to_3 + lambda_2_to_3
````

#### Expected Number of Entities in Stream 3 using Little's Law

Little's Law states that the average number of items in a stable system ($L$) is equal to the average arrival rate ($\lambda$) multiplied by the average time an item spends in the system ($W$).

For Stream 3:
-   Average Arrival Rate ($\lambda_{\text{total_effective_3}}$) = lambda_total_effective_3
-   Average Time in System ($W = d_3$) = duration_3

Thus, the expected number of entities in Stream 3 ($L_3$) is given by $L_3 = \lambda_{\text{total_effective_3}} \cdot d_3$.

````python
# Calculate expected number of entities in Stream 3 using Little's Law
expected_entities_stream3 = lambda_total_effective_3 * duration_3
````

#### Simulation to Verify Effective Arrival Rates into Stream 3

We will simulate the Poisson processes for Stream 1 and Stream 2, and the transition logic, to empirically verify the calculated effective arrival rates into Stream 3 ($\lambda_{1 \to 3}$, $\lambda_{2 \to 3}$, and $\lambda_{\text{total_effective_3}}$).

````python
import numpy as np
import random

def simulate_transitions(lambda_1, lambda_2, p_to_stream3_from1, p_to_stream3_from2, simulation_time):
    # Simulate departures from Stream 1 and Stream 2 based on their arrival rates
    # For this simulation, we consider that 'departures' from stream 1/2 are just events
    # occurring at the rate of the original stream.

    num_departures_s1 = np.random.poisson(lambda_1 * simulation_time)
    num_departures_s2 = np.random.poisson(lambda_2 * simulation_time)

    transitions_s1_to_s3 = 0
    for _ in range(num_departures_s1):
        if random.random() < p_to_stream3_from1:
            transitions_s1_to_s3 += 1

    transitions_s2_to_s3 = 0
    for _ in range(num_departures_s2):
        if random.random() < p_to_stream3_from2:
            transitions_s2_to_s3 += 1

    # Intrinsic arrivals to Stream 3
    intrinsic_s3_arrivals = np.random.poisson(lambda_3 * simulation_time)

    return transitions_s1_to_s3, transitions_s2_to_s3, intrinsic_s3_arrivals

# Set simulation parameters
simulation_duration = 1000 # hours
num_simulation_runs = 100 # number of times to run the simulation

empirical_lambda_1_to_3_sum = 0
empirical_lambda_2_to_3_sum = 0
empirical_lambda_s3_intrinsic_sum = 0

for _ in range(num_simulation_runs):
    t_s1, t_s2, t_s3_intrinsic = simulate_transitions(lambda_1, lambda_2, p_to_stream3_from1, p_to_stream3_from2, simulation_duration)
    empirical_lambda_1_to_3_sum += t_s1 / simulation_duration
    empirical_lambda_2_to_3_sum += t_s2 / simulation_duration
    empirical_lambda_s3_intrinsic_sum += t_s3_intrinsic / simulation_duration

empirical_lambda_1_to_3 = empirical_lambda_1_to_3_sum / num_simulation_runs
empirical_lambda_2_to_3 = empirical_lambda_2_to_3_sum / num_simulation_runs
empirical_lambda_s3_intrinsic = empirical_lambda_s3_intrinsic_sum / num_simulation_runs
empirical_lambda_total_effective_3 = empirical_lambda_s3_intrinsic + empirical_lambda_1_to_3 + empirical_lambda_2_to_3
````

The process of total arrivals is a Poisson process

A fundamental property of Poisson processes is that the superposition (sum) of two or more independent Poisson processes is also a Poisson process. The rate of the resulting process is the sum of the rates of the individual processes. ### Recap: Order Statistics of Three Independent Exponential Random Variables

Consider three independent streams, each starting at time 0. The lifetime of each stream is an independent and identically distributed (i.i.d.) Exponential random variable with rate parameter $\lambda$. Let $L_1, L_2, L_3 \sim \text{Exponential}(\lambda)$ be the lifetimes of the three streams.

We are interested in the ordered failure times:
- $T_1 = \min(L_1, L_2, L_3)$ (Time of the first stream to end)
- $T_2 = \text{median}(L_1, L_2, L_3)$ (Time of the second stream to end)
- $T_3 = \max(L_1, L_2, L_3)$ (Time of the third stream to end)

#### PDF of the Minimum $T_1$

Let $X_1 \sim Exp(\lambda_1)$, $X_2 \sim Exp(\lambda_2)$, and $X_3 \sim Exp(\lambda_3)$ be three independent exponential random variables. The CDF of the minimum $T_1$ is given by $F_{T_1}(t) = P(T_1 \le t) = 1 - P(T_1 > t)$

Given $T_1$ is the minimum, $T_1 > t$ if and only if all $X_i > t$. Due to independence $P(T_1 > t) = P(X_1 > t, X_2 > t, X_3 > t) = P(X_1 > t) P(X_2 > t) P(X_3 > t)$
We know that for an exponential distribution $X \sim Exp(\lambda)$, $P(X > t) = e^{-\lambda t}$. $P(T_1 > t) = e^{-\lambda_1 t} e^{-\lambda_2 t} e^{-\lambda_3 t} = e^{-(\lambda_1 + \lambda_2 + \lambda_3)t}$
Therefore, the CDF of $T_1$ is $F_{T_1}(t) = 1 - e^{-(\lambda_1 + \lambda_2 + \lambda_3)t}$, the CDF of an exponential distribution with rate parameter $(\lambda_1 + \lambda_2 + \lambda_3)$.

The PDF of $T_1$ is found by differentiating the CDF:
- $f_{T_1}(t) = \frac{d}{dt} F_{T_1}(t) = (\lambda_1 + \lambda_2 + \lambda_3) e^{-(\lambda_1 + \lambda_2 + \lambda_3)t}$ for $t \ge 0$.

**when $\lambda_1 = \lambda_2 = \lambda_3 = \lambda$:** If all rates are equal, the PDF simplifies to: $f_{T_1}(t) = (\lambda + \lambda + \lambda) e^{-(\lambda + \lambda + \lambda)t} = 3\lambda e^{-3\lambda t}$ for $t \ge 0$.

#### PDF of the Maximum $T_3$

Let $X_1 \sim Exp(\lambda_1)$, $X_2 \sim Exp(\lambda_2)$, and $X_3 \sim Exp(\lambda_3)$ be three independent exponential random variables. The CDF of the maximum $T_3$ is given by $F_{T_3}(t) = P(T_3 \le t)$

Given that $T_3$ is the maximum, $T_3 \le t$ if and only if all $X_i \le t$. Due to independence $P(T_3 \le t) = P(X_1 \le t, X_2 \le t, X_3 \le t) = P(X_1 \le t) P(X_2 \le t) P(X_3 \le t)$.
We know that for an exponential distribution $X \sim Exp(\lambda)$, $P(X \le t) = 1 - e^{-\lambda t}$. The CDF of $T_3$ is $F_{T_3}(t) = (1 - e^{-\lambda_1 t})(1 - e^{-\lambda_2 t})(1 - e^{-\lambda_3 t})$

To find the PDF, we differentiate $F_{T_3}(t)$ using the product rule. Let $A = (1 - e^{-\lambda_1 t})$, $B = (1 - e^{-\lambda_2 t})$, $C = (1 - e^{-\lambda_3 t})$. Then $F_{T_3}(t) = ABC$, and $f_{T_3}(t) = A'BC + AB'C + ABC'$.
$A' = \lambda_1 e^{-\lambda_1 t}$, $B' = \lambda_2 e^{-\lambda_2 t}$, $C' = \lambda_3 e^{-\lambda_3 t}$. Substituting these, we get:
- $f_{T_3}(t) = \lambda_1 e^{-\lambda_1 t} (1 - e^{-\lambda_2 t})(1 - e^{-\lambda_3 t})
\quad + (1 - e^{-\lambda_1 t}) \lambda_2 e^{-\lambda_2 t} (1 - e^{-\lambda_3 t})
\quad + (1 - e^{-\lambda_1 t})(1 - e^{-\lambda_2 t}) \lambda_3 e^{-\lambda_3 t}$ for $t \ge 0$.

**when $\lambda_1 = \lambda_2 = \lambda_3 = \lambda$:** If all rates are equal, the CDF simplifies to: $F_{T_3}(t) = (1 - e^{-\lambda t})^3$. Expanding this, we get $F_{T_3}(t) = 1 - 3e^{-\lambda t} + 3e^{-2\lambda t} - e^{-3\lambda t}$.

Differentiating to find the PDF
- $f_{T_3}(t) = 3\lambda e^{-\lambda t} - 6\lambda e^{-2\lambda t} + 3\lambda e^{-3\lambda t}$ for $t \ge 0$.

#### PDFs for Order Statistics of Three Independent Exponential Random Variables

Let $X_1 \sim Exp(\lambda_1)$, $X_2 \sim Exp(\lambda_2)$, and $X_3 \sim Exp(\lambda_3)$ be three independent exponential random variables. The PDF of an exponential distribution is $f(x; \lambda) = \lambda e^{-\lambda x}$ for $x \ge 0$, and the CDF is $F(x; \lambda) = 1 - e^{-\lambda x}$ for $x \ge 0$.

#### PDF of the Median (median($X_1, X_2, X_3$))

Let's denote the order statistics as $X_{(1)} \le X_{(2)} \le X_{(3)}$, where $X_{(1)}$ is the minimum, $X_{(2)}$ is the median, and $X_{(3)}$ is the maximum. We are looking for the PDF of $X_{(2)}$.

For three independent exponential random variables $X_1 \sim Exp(\lambda_1)$, $X_2 \sim Exp(\lambda_2)$, and $X_3 \sim Exp(\lambda_3)$, with PDFs $f_i(x) = \lambda_i e^{-\lambda_i x}$ and CDFs $F_i(x) = 1 - e^{-\lambda_i x}$, the PDF of the median $X_{(2)}$ is given by the general formula for the $k$-th order statistic for $n$ independent variables. For $n=3$ and $k=2$ (the median), the pdf simplifies to
- $f_{X_{(2)}}(x) = f_1(x)[F_2(x)(1-F_3(x)) + (1-F_2(x))F_3(x)] + f_2(x)[F_1(x)(1-F_3(x)) + (1-F_1(x))F_3(x)] + f_3(x)[F_1(x)(1-F_2(x)) + (1-F_1(x))F_2(x)]$

Let's break down this expression and substitute the exponential PDF and CDF components: $f_i(x) = \lambda_i e^{-\lambda_i x}$, $F_i(x) = 1 - e^{-\lambda_i x}$, $(1-F_i(x)) = e^{-\lambda_i x}$.

Substituting these into the pdf for $f_{X_{(2)}}(x)$ and
expanding the terms we get the full PDF of the median $X_{(2)}$
- $f_{X_{(2)}}(x) = \lambda_1 e^{-\lambda_1 x} (e^{-\lambda_2 x} + e^{-\lambda_3 x} - 2e^{-(\lambda_2+\lambda_3)x}) + \lambda_2 e^{-\lambda_2 x} (e^{-\lambda_1 x} + e^{-\lambda_3 x} - 2e^{-(\lambda_1+\lambda_3)x}) + \lambda_3 e^{-\lambda_3 x} (e^{-\lambda_1 x} + e^{-\lambda_2 x} - 2e^{-(\lambda_1+\lambda_2)x})$

This expression can be further expanded and collected, but this form clearly shows the contributions from each exponential variable and their combinations. It is valid for $x \ge 0$.


##### PDF of the Median (median($X_1, X_2, X_3$)) when $\lambda_1 = \lambda_2 = \lambda_3 = \lambda$

If $X_1, X_2, X_3 \sim Exp(\lambda)$, we substitute $\lambda_1 = \lambda_2 = \lambda_3 = \lambda$ into the general formula for $f_{X_{(2)}}(x)$:

$f_{X_{(2)}}(x) = \lambda e^{-\lambda x} (e^{-\lambda x} + e^{-\lambda x} - 2e^{-(\lambda+\lambda)x}) + \lambda e^{-\lambda x} (e^{-\lambda x} + e^{-\lambda x} - 2e^{-(\lambda+\lambda)x}) + \lambda e^{-\lambda x} (e^{-\lambda x} + e^{-\lambda x} - 2e^{-(\lambda+\lambda)x})$

Simplifying the term in the parentheses for each part:
$(e^{-\lambda x} + e^{-\lambda x} - 2e^{-2\lambda x}) = (2e^{-\lambda x} - 2e^{-2\lambda x})$

- $f_{X_{(2)}}(x) = 6 \lambda (e^{-2\lambda x} - e^{-3\lambda x})$ for $x \ge 0$.


### Case Study: Ordered Failures in Three Independent Exponential Streams

Consider three independent streams, each starting at time 0. The lifetime of each stream is an independent and identically distributed (i.i.d.) Exponential random variable with rate parameter $\lambda$. Let $L_1, L_2, L_3 \sim \text{Exponential}(\lambda)$ be the lifetimes of the three streams.

We are interested in the ordered failure times:
- $T_1 = \min(L_1, L_2, L_3)$ (Time of the first stream to end)
- $T_2 = \text{median}(L_1, L_2, L_3)$ (Time of the second stream to end)
- $T_3 = \max(L_1, L_2, L_3)$ (Time of the third stream to end)

We will derive the distributions for $T_1$, $T_2$, and $T_3$, and then focus on the differences between these ordered statistics.

##### Distributions of $T_1$, $T_2$, and $T_3$

**1. Distribution of $T_1 = \min(L_1, L_2, L_3)$**

Since $L_1, L_2, L_3$ are i.i.d. $\text{Exponential}(\lambda)$, their minimum is also Exponential: $P(T_1 > t) = P(L_1 > t, L_2 > t, L_3 > t) = P(L_1 > t)P(L_2 > t)P(L_3 > t)$ (due to independence), $P(T_1 > t) = (e^{-\lambda t})^3 = e^{-3\lambda t}$. Therefore, $T_1 \sim \text{Exponential}(3\lambda)$.

**2. Distribution of $T_3 = \max(L_1, L_2, L_3)$**

The maximum of i.i.d. Exponential random variables does not follow a simple Exponential distribution. Its CDF is:
$P(T_3 \le t) = P(L_1 \le t, L_2 \le t, L_3 \le t) = P(L_1 \le t)P(L_2 \le t)P(L_3 \le t)$ (due to independence)
$P(T_3 \le t) = (1 - e^{-\lambda t})^3$.

**3. Distribution of $T_2 = \text{median}(L_1, L_2, L_3)$**

The distribution of the median is more complex to derive directly. However, we can leverage properties of order statistics for i.i.d. Exponential random variables in the context of Poisson processes. The inter-arrival times between ordered failures are also exponentially distributed, but with changing rates due to the memoryless property.

##### Conditional PDF for $X = T_2 - T_1$ given $T_1 = t_1$

Let $X = T_2 - T_1$ be the additional time from the first failure until the second failure. We want to find $f_{X|T_1}(x|t_1)$ for $x, t_1 > 0$. Given $T_1 = t_1$, one of the three streams has failed at time $t_1$. Due to the memoryless property of the Exponential distribution, the remaining two streams are still alive and their remaining lifetimes are i.i.d. Exponential($\lambda$).
The time until the next failure among these two remaining streams is $X$. Thus, $X$ is the minimum of two i.i.d. Exponential($\lambda$) random variables.
Therefore, $X \sim \text{Exponential}(2\lambda)$.
The conditional PDF $f_{X|T_1}(x|t_1)$ is simply the PDF of an Exponential($2\lambda$) distribution, as it is independent of $t_1$ due to memorylessness:
$f_{X|T_1}(x|t_1) = 2\lambda e^{-2\lambda x}$ for $x > 0$.

##### Conditional PDF for $Y = T_3 - T_2$ given $T_2 = t_2$

Let $Y = T_3 - T_2$ be the additional time from the second failure until the third (last) failure. We want to find $f_{Y|T_2}(y|t_2)$ for $y, t_2 > 0$.

Given $T_2 = t_2$, two of the three streams have failed by time $t_2$. Due to the memoryless property, the *one* remaining stream is still alive and its remaining lifetime is Exponential($\lambda$).

The time until this last remaining stream fails is $Y$. Thus, $Y$ is an Exponential($\lambda$) random variable.

Therefore, $Y \sim \text{Exponential}(\lambda)$.

The conditional PDF $f_{Y|T_2}(y|t_2)$ is the PDF of an Exponential($\lambda$) distribution, also independent of $t_2$:
$f_{Y|T_2}(y|t_2) = \lambda e^{-\lambda y}$ for $y > 0$.

### Applications and Use Cases

These probability concepts have various interesting applications in real-world scenarios:

**1. Discrete Probability (Item Delivery) problems:**

*   **Inventory Management & Logistics**: Predicting when a new item will be needed, or when a delivery system might run out of stock. For example, if you're a delivery service, how likely are you to run out of a certain package type by a certain stop number?
*   **Quality Control/Inspections**: If defects occur with a certain probability per unit inspected, these calculations can help determine the likelihood of finding the *k*-th defect by a specific inspection point, or when a new batch of items might be needed.
*   **Customer Service/Interaction Models**: Modeling the probability of a customer needing a certain type of support after a series of interactions.
*   **Gaming/Chance**: Calculating probabilities of specific events occurring in turn-based games or other scenarios with discrete steps and success/failure outcomes.

**2. Continuous Probability (Order Statistics of Exponential Distributions) problems:**

*   **Reliability Engineering**: Exponential distributions are often used to model the lifetime of components in a system (assuming a constant failure rate).
    *   **Minimum**: If you have multiple redundant components (e.g., three backup servers), the system fails when the *first* component fails. The PDF of the minimum helps predict when the overall system will fail.
    *   **Maximum**: If a system operates as long as *at least one* of its components is functional (e.g., three identical light bulbs in a single fixture, the light goes out when the *last* bulb burns out), the PDF of the maximum helps predict the system's total operational lifetime.
    *   **Median**: In more complex systems with multiple components, understanding the median lifetime can be useful for maintenance scheduling or predicting the 'typical' time for half of a batch of components to fail.
*   **Queuing Theory**: Analyzing waiting times or service times in systems like call centers, computer networks, or checkout lines. Order statistics can describe, for instance, the time until the first customer is served, or the time until the last customer leaves a small queue.
*   **Telecommunications**: Modeling call durations or the time until the first or last of several concurrent data streams completes.
*   **Survival Analysis (Medical/Biological Studies)**: Estimating the time until a particular event occurs (e.g., disease recurrence, patient response to treatment) across a population, where individual times might be modeled exponentially. Order statistics help analyze the distribution of these event times.
