#### Case Study: Store Closing Based on Actual Customers
A store opens at t = 0 and potential customers arrive in a Poisson manner at an average arrival rate of $\lambda$ potential customers per hour. As long as the store is open, and independently of all other events, each particular potential customer becomes an actual customer with probability $p$.
The store closes as soon as ten actual customers have arrived.

(a) What is the probability that exactly three of the first five potential customers become actual customers?
(b) What is the probability that the fifth potential customer to arrive becomes the third actual customer?
(c) What is the PDF and expected value for L, the duration of the interval from store opening to store closing?
(d) Given only that exactly three of the first five potential customers became actual customers, what is the conditional expected value of the total time the store is open?
(e) Considering only customers arriving between t = 0 and the closing of the store, what is the
probability that no two actual customers arrive within $\tau$ time units of each other?

#### Parameters for the Case Study

Let's define the given parameters for our store scenario.

```python
import math
import numpy as np

# Given parameters
lambda_potential = 10.0  # Average arrival rate of potential customers per hour
p = 0.6               # Probability a potential customer becomes an actual customer
num_actual_to_close = 10 # Number of actual customers to close the store
```

##### (a) What is the probability that exactly three of the first five potential customers become actual customers?

Since a given potential customer becomes an actual customer with probability $p$, we have a binomial probability. The probability of exactly 3 of the first 5 potential customers being actual customers is $\text{Comb}(5, 3) \cdot p^3 \cdot (1 - p)^2$.

```python
# Part (a) Calculation
N_trials_a = 5
k_successes_a = 3
prob_part_a = math.comb(N_trials_a, k_successes_a) * (p**k_successes_a) * ((1 - p)**(N_trials_a - k_successes_a))
```

##### (b) What is the probability that the fifth potential customer to arrive becomes the third actual customer?

Potential customers become actual customers independently of each other.
$P(\text{the fifth customer is the third actual customer}) = P(\text{any 2 of the first 4 customers become actual customer}) \cdot P(\text{the fifth customer becomes an actual customer})$
$= (\text{Comb}(4, 2) \cdot p^2 \cdot (1 - p)^2) \cdot p = (\text{Comb}(4, 2) \cdot p^3 \cdot (1 - p)^2)$

```python
# Part (b) Calculation
# P(2 of first 4 are actual) * P(5th is actual)
prob_first_four_b = math.comb(4, 2) * (p**2) * ((1 - p)**2)
prob_fifth_is_actual_b = p
prob_part_b = prob_first_four_b * prob_fifth_is_actual_b
```

##### (c) What is the PDF and expected value for L, the duration of the interval from store opening to store closing?

The process of incoming customers is randomly split into two independent Poisson processes: the process for the arrival of actual customers, with rate $p \cdot \lambda$, and the process for the arrival of potential customers who do not become actual customers, with rate $(1 - p) \cdot \lambda$.
The store will close when 10 actual customers have arrived. Thus, the store closes when we have 10 arrivals from the Poisson process of actual customers, which has rate $p \cdot \lambda$. Thus, $L = T_1 + T_2 + \dots + T_{10}$, where $T_i$ is the interarrival time with an exponential distribution and parameter $p \cdot \lambda$. The distribution of $L$ is the 10th order Erlang PDF, $f_L(l) = \frac{(p \cdot \lambda)^{10} l^9 e^{-(p \cdot \lambda)l}}{9!}$ for $l \ge 0$.
It follows that the expected value of $L$ is $\frac{10}{p \cdot \lambda}$.

```python
# Part (c) Calculation
lambda_actual_c = lambda_potential * p
k_erlang_c = num_actual_to_close

pdf_L_c_str = f"f_L(l) = ({lambda_actual_c:.2f})^{{k_erlang_c}} l^{{k_erlang_c-1}} e^{{-{{lambda_actual_c:.2f}} l}} / {math.factorial(k_erlang_c-1)}!  for l >= 0"
expected_L_c = k_erlang_c / lambda_actual_c
```

##### (d) Given only that exactly three of the first five potential customers became actual customers, what is the conditional expected value of the total time the store is open?

Since five potential customers have arrived, three of which are actual customers, we are interested in the time for the next seven actual customers to arrive. Following from part (c), the expected time for the next seven actual customers to arrive is $\frac{7}{p \cdot \lambda}$. Adding the expected time for the first five potential customers to arrive, we get that the conditional expectation for the total time the store is open is $\frac{5}{\lambda} + \frac{7}{p \cdot \lambda}$.

```python
# Part (d) Calculation
lambda_actual_d = lambda_potential * p

# Expected time for the first 5 potential customers to arrive
expected_T5_potential = 5 / lambda_potential

# Expected time for the remaining 7 actual customers (10 total - 3 already arrived)
expected_additional_7_actual = (num_actual_to_close - 3) / lambda_actual_d

conditional_expected_L_d = expected_T5_potential + expected_additional_7_actual
```

##### (e) Considering only customers arriving between t = 0 and the closing of the store, what is the probability that no two actual customers arrive within $\tau$ time units of each other?

The probability of no two actual customers arriving within $\tau$ time units of each other is equivalent to the probability of all nine independent interarrival times, separating the ten actual customers, being at least $\tau$ time units apart. Thus, $P(\text{no two actual customers arriving within } \tau \text{ time units of each other}) = P([T_i \ge \tau]^9) = e^{-9 \cdot p \cdot \lambda \cdot \tau}$.

Let's choose an example value for $\tau$, say $\tau = 0.05$ hours.

```python
# Part (e) Calculation
tau_e = 0.05 # Example value for tau in hours
lambda_actual_e = lambda_potential * p

# As per the provided solution, considering 9 inter-arrival times
num_inter_arrivals_e = num_actual_to_close - 1 # 9 inter-arrival times

prob_part_e = np.exp(-num_inter_arrivals_e * lambda_actual_e * tau_e)
```

#### Case Study:  Bank Tellers

When you enter your bank, you find that there are only two tellers, both busy serving other customers, and that there are no other customers in line. Assume that the service times for you and for each of the customers being served are iid rvs with parameter $\mu$. Also assume that after a service completion, the next customer in line immediately begins to be served.

1.  Expectation from your arrival until 'k' customers are finished being served.
2.  Probability that you will be served last using the M/M/2 model.

#### Derivation: Expectation from your arrival until 'k' customers are finished being served

Let $S_A$ and $S_B$ be the service times of the two customers initially being served, and $S_Y$ be your service time. All i.i.d. exponential random variables with rate $\mu$. You arrive when both tellers are busy, and there are no other customers in line.

Let $E[T_k]$ denote the expected time from your arrival until $k$ customers (from the initial state) have finished being served.

1.  **Time until 1st service completion ($E[T_1]$):**
    Upon your arrival, two customers ($S_A, S_B$) are being served. The time until the first service completion is $T_1 = \min(S_A, S_B)$. Since $S_A, S_B \sim \text{Exponential}(\mu)$, then $T_1 \sim \text{Exponential}(2\mu)$.
    $E[T_1] = \frac{1}{2\mu}$.

    At this point, one customer finishes, and one teller becomes free. You immediately move to this teller and begin your service. Now, two customers are being served: the remaining original customer (with remaining service time $S_B' \sim \text{Exponential}(\mu)$ due to memorylessness) and you ($S_Y \sim \text{Exponential}(\mu)$).

2.  **Time until 2nd service completion ($E[T_2]$):**
    After $T_1$, two customers ($S_B'$ and $S_Y$) are being served concurrently. The time until the next service completion (which will be the second overall completion since your arrival) is $T_{add_2} = \min(S_B', S_Y) \sim \text{Exponential}(2\mu)$.
    $E[T_2] = E[T_1] + E[T_{add_2}] = \frac{1}{2\mu} + \frac{1}{2\mu} = \frac{1}{\mu}$.

    At this point, two customers have finished service since your arrival. One teller is now free. One customer (either $S_B'$ or you) is still being served.

3.  **Time until 3rd service completion ($E[T_3]$):**
    After $T_2$, two customers have left, and one customer (the one who was still being served) is completing their service. The other teller is idle. The time until this third service completion is $T_{add_3} \sim \text{Exponential}(\mu)$.
    $E[T_3] = E[T_2] + E[T_{add_3}] = \frac{1}{\mu} + \frac{1}{\mu} = \frac{2}{\mu}$.

4.  **Generalizing for $k$ customers finished ($E[T_k]$):**
    *   For the first two completions (i.e., $k=2$), the total expected time is $E[T_2] = \frac{1}{\mu}$.
    *   For any subsequent completions (i.e., $k > 2$), only one customer is being served at a time (either you, or the next customer in line if you have finished). Each additional service completion takes an average of $\frac{1}{\mu}$ time.

    Therefore, for $k \ge 2$:
    $E[T_k] = E[T_2] + (k-2) \cdot \frac{1}{\mu} = \frac{1}{\mu} + \frac{k-2}{\mu} = \frac{k-1}{\mu}$.

**Summary:**
*   $E[T_1] = \frac{1}{2\mu}$
*   $E[T_k] = \frac{k-1}{\mu}$ for $k \ge 2$

#### Derivation: Probability that you will be served last using the M/M/2 model

Let the two customers initially being served be Customer 1 ($C_1$) and Customer 2 ($C_2$). You are Customer Y ($C_Y$). All service times ($S_1, S_2, S_Y$) are i.i.d. Exponential random variables with rate $\mu$.

You are served last if both $C_1$ and $C_2$ complete their service before you do.

1.  **First Service Completion:** When you arrive, $C_1$ and $C_2$ are being served. The time until one of them finishes is $T_A = \min(S_1, S_2)$. Due to the symmetry and i.i.d. nature of exponential service times, $C_1$ finishes first with probability 1/2, and $C_2$ finishes first with probability 1/2.

2.  **Your Entry into Service:** As soon as one of the initial customers (say $C_1$) finishes, you immediately move to that teller and begin your service. At this moment, $C_2$ is still being served (with a remaining service time $S_2' \sim \text{Exponential}(\mu)$ due to memorylessness), and you ($C_Y$) have just started service ($S_Y \sim \text{Exponential}(\mu)$).

3.  **The "Race" between You and the Remaining Customer:** Now, for you to be served last, $C_2$ (the remaining original customer) must finish their service *before* you finish yours. We are comparing $S_2'$ and $S_Y$.

    For two independent Exponential random variables $X \sim \text{Exponential}(\lambda_X)$ and $Y \sim \text{Exponential}(\lambda_Y)$, the probability that $X$ finishes before $Y$ is given by $P(X < Y) = \frac{\lambda_X}{\lambda_X + \lambda_Y}$.

    In our case, $X = S_2'$ (remaining service for $C_2$) and $Y = S_Y$ (your service). Both have a rate of $\mu$.

    Therefore, the probability that $S_2'$ finishes before $S_Y$ is:
    $P(S_2' < S_Y) = \frac{\mu}{\mu + \mu} = \frac{\mu}{2\mu} = \frac{1}{2}$.

This outcome is independent of which of $C_1$ or $C_2$ finished first. In any scenario where one of the initial customers finishes, you move into service alongside the other, and the race between you and that other customer to finish next has a 1/2 probability of that other customer winning. If that other customer wins, you are served last (as the first customer is already gone).

Thus, the probability that you will be served last is $\frac{1}{2}$.

#### Analyzing the Bank Tellers Problem through Stochastic Processes (M/M/2 Queuing Model)

The bank tellers problem can be formally modeled as a **Continuous-Time Markov Chain (CTMC)**, and more specifically, as an **M/M/c queuing system** where $c=2$ (two tellers). Let's break down how this model applies:

**M/M/c Notation:**
*   **M (Markovian Arrivals):** This part usually refers to Poisson arrivals. In our specific problem, we are not explicitly given an arrival process for *new* customers after the initial two. However, the problem focuses on the service times, which are indeed Markovian (Exponentially distributed).
*   **M (Markovian Service Times):** This is explicitly stated. The service times for both tellers and yourself are i.i.d. Exponential random variables with rate $\mu$. The memoryless property of the Exponential distribution is crucial here, as it means the remaining service time for a customer is independent of how long they have already been served.
*   **c (Number of Servers):** In this case, there are $c=2$ tellers, acting as servers.

**State Definition:**
We can define the state of the system based on the number of customers being served. However, since we are interested in specific customers (the two initial ones and yourself), a more detailed state description focuses on *who* is being served.

**Key Events and Transition Rates:**
1.  **Initial State:** At your arrival, two customers ($C_1, C_2$) are being served. Both tellers are busy. No queue. The effective service rate for the system is $2\mu$ (since two services are happening in parallel).
2.  **First Service Completion:** The time until the first of $C_1$ or $C_2$ finishes service is $\text{Exponential}(2\mu)$. When one finishes (e.g., $C_1$), the system transitions.
3.  **Your Service Begins:** Immediately after the first service completion, you (Customer Y) begin service at the newly free teller. Now, two customers are being served again: the remaining initial customer ($C_2'$) and you ($C_Y$). The effective service rate for the system is again $2\mu$.
4.  **Second Service Completion:** The time until either $C_2'$ or $C_Y$ finishes service is $\text{Exponential}(2\mu)$. When one finishes, the system transitions.
5.  **Subsequent Service Completions:** After two service completions (meaning both $C_1$ and $C_2$ are gone, and you are either still being served or have finished, and a new customer might be served), only one customer will be served at a time unless another customer was waiting. Given no other customers in line, any subsequent service completion involves only one busy teller, so the effective service rate is $\mu$.

**Relation to our Calculations:**
*   **Expected Time to $k$ Completions:** Our derivations for $E[T_k]$ explicitly use the rates $2\mu$ (when two tellers are busy) and $\mu$ (when only one teller is busy), which directly comes from the CTMC formulation. For instance, the time until the first service completion being $1/(2\mu)$ is a direct result of the minimum of two Exponential random variables with rate $\mu$.
*   **Probability of Being Served Last:** The memoryless property of the Exponential distribution is critical here. When one of the initial customers finishes, their remaining service time is 'reset' due to the memoryless property. The subsequent 'race' between you and the other remaining initial customer being served is a classic application of comparing two independent Exponential random variables, leading to the $1/2$ probability.

This stochastic process view provides a rigorous foundation for the calculations performed, highlighting the role of Exponential distributions and their properties in queuing theory.

### Classic Problems with Poisson Processes

#### Arrival of Customers

This is a cornerstone application of Poisson processes. Imagine a bank, a supermarket checkout, or a call center. The number of customers or clients arriving per unit of time (e.g., per hour, per minute) can often be effectively modeled by a Poisson process. The core assumptions are that each arrival is an independent event, and the average rate of arrivals remains constant over the period of observation. This model is invaluable for operations research, helping businesses determine optimal staffing levels, predict queue lengths, manage customer waiting times, and improve overall service efficiency. By understanding the arrival patterns, resources can be allocated more effectively to meet demand.

Specifics:
*   **Queueing Theory:** Poisson arrivals are fundamental in M/M/1, M/M/c, and other queueing models for predicting wait times and system utilization.
*   **Staffing Optimization:** Businesses use this to forecast demand and schedule staff to minimize both customer wait times and labor costs.
*   **Traffic Flow:** Can model the arrival of vehicles at an intersection or toll booth.
*   **Call Centers:** Predicting the number of incoming calls for capacity planning and agent scheduling.

##### Expressions/Formulas for Customer Arrivals

If the average rate of customer arrivals is \(\lambda\) customers per hour, the probability of exactly \(k\) customers arriving in a given hour is given by the Poisson PMF $[P(X=k) = \frac{e^{-\lambda} \lambda^k}{k!}]$
To find the probability of _at most_ $(k)$ customers, we'd sum the probabilities from 0 to $(k)$ (CDF): $(P(X \le k) = \sum_{i=0}^{k} \frac{e^{-\lambda} \lambda^i}{i!})$.

#### Phone Calls

Similar to customer arrivals, the number of incoming phone calls to a large call center is a classic example of a Poisson process. During periods of stable demand, individual calls are largely independent, and the rate at which they arrive can be considered constant. This model is crucial for telecommunication companies and call centers to predict traffic loads, optimize the number of agents needed, plan network capacity, and minimize dropped calls or long wait times. It helps in understanding the probability of receiving a certain number of calls within a specific interval, which directly impacts service quality and cost.

Specifics:
*   **Telecommunications:** Designing network infrastructure, allocating bandwidth, and managing trunk lines based on expected call volumes.
*   **Emergency Services (e.g., 911):** Predicting peak call times to ensure sufficient operators and resources are available.
*   **Marketing Campaign Response:** Estimating the number of calls generated by a specific advertising campaign.

##### Expressions/Formulas for Phone Calls

If a call center receives an average of $(\lambda)$ calls per minute, the probability of receiving exactly $(k)$ calls in a specific minute is
$[P(X=k) = \frac{e^{-\lambda} \lambda^k}{k!}]$.
This can be used to determine the likelihood of high call volumes to ensure adequate staffing.
