### Poisson Processes

A Poisson process is a stochastic process that models the occurrence of events (arrivals) at a constant average rate, independently of previous events. It's often used to describe phenomena such as customer arrivals at a store, phone calls to a call center, radioactive decay events, or network traffic.

Definition and Applications

A key characteristic of a Poisson process is that the number of arrivals in disjoint time intervals are independent. This property simplifies the analysis of such processes.

Let $P(k, \tau)$ be the probability of $k$ arrivals in a time interval of duration $\tau$. For a Poisson process with an average arrival rate $\lambda$ (lambda) per unit time:

**Small Interval Probabilities**: For a very small time interval $\delta t$
-   The probability of **zero arrivals** is approximately $P(0, \delta t) = 1 - \lambda \cdot \delta t + O((\delta t)^2)$.
-   The probability of **one arrival** is approximately $P(1, \delta t) = \lambda \cdot \delta t + O((\delta t)^2)$.
-   The probability of **more than one arrival** (i.e., $k > 1$) is approximately: $P(k > 1, \delta t) = 0 + O((\delta t)^2)$.

Where $O((\delta t)^2)$ represents terms that become negligible as $\delta t$ approaches zero. For practical purposes, when $\delta t$ is sufficiently small, these can be simplified to
-   $P(0, \delta t) \approx 1 - \lambda \cdot \delta t$, $\quad$ $P(1, \delta t) \approx \lambda \cdot \delta t$, $\quad$ $P(k > 1, \delta t) \approx 0$

Here, $\lambda$ is the **arrival rate**, representing the average number of events per unit time.

#### Distribution of Number of Arrivals (Poisson PMF)

For a Poisson process with arrival rate $\lambda$, the number of arrivals in a time interval of duration $\tau$, denoted as $N_\tau$, follows a Poisson distribution.

This can be understood by approximating the continuous time interval with $n$ very small discrete intervals (slots) of length $\delta = \tau/n$. In each small slot, the probability of one arrival is approximately $\lambda \delta$, and the probability of more than one arrival is negligible ($O((\delta t)^2)$).

Thus, $N_\tau$ can be approximated by a Binomial distribution $\text{Binomial}(n, p)$, where
- $n = \tau / \delta$ (number of trials), $p = \lambda \delta$ (probability of success in each trial)

The mean of this Binomial distribution is $n \cdot p = (\tau/\delta) \cdot (\lambda \delta) = \lambda \tau$. As $\delta \to 0$ (and thus $n \to \infty$), the Binomial distribution converges to the Poisson distribution with parameter $\lambda \tau$.

The pmf for the number of arrivals $k$ in a time interval of duration $\tau$ is given by
- $P(N_\tau = k) = \frac{(\lambda\tau)^k e^{-\lambda\tau}}{k!}$, for $k = 0, 1, 2, \dots$, where
$N_\tau$ is the number of arrivals in the interval $[0, \tau]$, $\lambda$ is the average arrival rate per unit time, $\tau$ is the duration of the time interval, $k$ is the observed number of arrivals.

**Key properties of $N_\tau$** Expectation $E[N_\tau] = \sum_{k=0}^{\infty} k \cdot P(N_\tau = k) = \sum_{k=0}^{\infty} k \frac{(\lambda\tau)^k e^{-\lambda\tau}}{k!} = \lambda\tau$, Variance $Var(N_\tau) = \lambda \tau$.

These can also be derived from the Binomial approximation $N_\tau \sim \text{Binomial}(n, p)$, where $n = \tau/\delta$ and $p = \lambda\delta$. Then $E[N_\tau] = np = (\tau/\delta)(\lambda\delta) = \lambda\tau$, and $Var(N_\tau) = np(1-p) \approx np = \lambda\tau$ (as $p \to 0$ for small $\delta$).

#### Email Arrivals (Poisson Process)

Consider a scenario where you receive emails according to a Poisson process at a rate of $\lambda = 5$ messages per hour.

-   **Mean and Variance of emails received during a day (24 hours)**:
For a time interval $\tau = 24$ hours, the parameter for the Poisson distribution is $\lambda \tau = 5 \text{ messages/hour} \times 24 \text{ hours} = 120$. Therefore Mean $E[N_{24}] = \lambda \tau = 120$ and Variance $Var(N_{24}) = \lambda \tau = 120$.

-   **Probability of exactly one new message in the next hour**:
Here, $\tau = 1$ hour. We want to find $P(N_1 = 1)$. Using the Poisson PMF $P(N_1 = 1) = \frac{(\lambda\tau)^1 e^{-\lambda\tau}}{1!} = \frac{(5 \times 1)^1 e^{-(5 \times 1)}}{1!} = 5 e^{-5}$.

-   **Probability of exactly two messages during each of the next three hours**:
    Since arrivals in disjoint time intervals are independent for a Poisson process, we can calculate the probability for one hour and raise it to the power of three.
    For one hour ($\tau=1$), the probability of exactly two messages is:
    $P(N_1 = 2) = \frac{(\lambda\tau)^2 e^{-\lambda\tau}}{2!} = \frac{(5 \times 1)^2 e^{-(5 \times 1)}}{2!} = \frac{25 e^{-5}}{2}$.
    
    So, the probability of exactly two messages during each of the next three hours is:
    $(P(N_1 = 2))^3 = \left(\frac{25 e^{-5}}{2}\right)^3$

#### Time until the First Arrival ($T_1$ - Exponential Distribution)

For a Poisson process with arrival rate $\lambda$, the time until the first arrival, $T_1$, follows an Exponential distribution.

-   The probability of $k$ arrivals in a time interval of duration $\tau$ is given by the Poisson pmf,
$P(N_\tau = k) = \frac{(\lambda\tau)^k e^{-\lambda\tau}}{k!}$.

-   The CDF for $T_1$ can be derived by considering the probability that the first arrival occurs by time $t$:
$P(T_1 \le t) = 1 - P(T_1 > t)$
$P(T_1 > t)$ is the probability that there are zero arrivals in the interval $(0, t]$, which is $P(N_t = 0)$.
$P(N_t = 0) = \frac{(\lambda t)^0 e^{-\lambda t}}{0!} = e^{-\lambda t}$
Therefore, $P(T_1 \le t) = 1 - e^{-\lambda t}$ for $t \ge 0$.

-   The pdf of $T_1$ is found by differentiating the CDF,
$f_{T_1}(t) = \frac{d}{dt} (1 - e^{-\lambda t}) = \lambda e^{-\lambda t}$ for $t \ge 0$.
This is the PDF of an **Exponential distribution** with parameter $\lambda$.

-   **Memorylessness Property**: The Exponential distribution (and thus the time until the next arrival in a Poisson process) exhibits the memorylessness property. This means that the probability distribution of the *additional* time until the next arrival does not depend on how much time has already passed without an arrival. Mathematically, for an Exponential random variable $T_1$,
$P(T_1 > t+s \mid T_1 > t) = P(T_1 > s)$ for any $t, s \ge 0$.
In other words, if no arrival has occurred by time $t$, the distribution of the remaining time until the next arrival is the same as the original distribution of $T_1$, starting afresh from time $t$.

#### Time of the k-th Arrival ($Y_k$ - Erlang Distribution)

For a Poisson process with arrival rate $\lambda$, the time until the $k$-th arrival, $Y_k$, follows an **Erlang distribution** with parameters $k$ (shape parameter, number of arrivals) and $\lambda$ (rate parameter).

-   **Deriving the CDF**: The probability that the $k$-th arrival occurs by time $y$, $P(Y_k \le y)$, is equivalent to the probability that the number of arrivals in the interval $(0, y]$ is greater than or equal to $k$.

    $P(Y_k \le y) = P(N_y \ge k) = \sum_{n=k}^{\infty} P(N_y = n) = \sum_{n=k}^{\infty} \frac{(\lambda y)^n e^{-\lambda y}}{n!}$

-   **Deriving the PDF**: The probability density function $f_{Y_k}(y)$ can be found by differentiating the CDF, or more intuitively, by considering the probability of the $k$-th arrival occurring in a small interval $(y, y+\delta]$.

    The event that $Y_k \in (y, y+\delta]$ means that there were exactly $k-1$ arrivals in $(0, y]$ and exactly one arrival in $(y, y+\delta]$.
    
    So, $f_{Y_k}(y) \cdot \delta \approx P(N_y = k-1 \text{ and } N_{(y, y+\delta]} = 1)$
    Due to the independence of arrivals in disjoint intervals:
    $f_{Y_k}(y) \cdot \delta \approx P(N_y = k-1) \cdot P(N_{(y, y+\delta]} = 1)$
    $f_{Y_k}(y) \cdot \delta \approx \left(\frac{(\lambda y)^{k-1} e^{-\lambda y}}{(k-1)!}\right) \cdot (\lambda \delta + O((\delta)^2))$

    Dividing by $\delta$ and taking the limit as $\delta \to 0$, we get the PDF:
    $f_{Y_k}(y) = \frac{\lambda^k y^{k-1} e^{-\lambda y}}{(k-1)!}$ for $y \ge 0$.

-   **Expectation and Variance**: For an Erlang($k, \lambda$) distribution:
    -   $E[Y_k] = \frac{k}{\lambda}$, $Var(Y_k) = \frac{k}{\lambda^2}$

This makes sense because $Y_k$ is the sum of $k$ independent and identically distributed Exponential random variables, each with mean $1/\lambda$ and variance $1/\lambda^2$. The sum of $k$ such variables would have mean $k/\lambda$ and variance $k/\lambda^2$.

#### Memorylessness and Fresh-Start Properties in Poisson Processes

Poisson processes exhibit strong memorylessness and fresh-start properties, which are crucial for understanding their behavior:

-   **Fresh-Start Property from any point $t$**: If we observe a Poisson process with rate $\lambda$ up to time $t$, and then start observing it from time $t$ onwards, the process *from time $t$* behaves exactly like a new, independent Poisson process with the same rate $\lambda$. This means that the future behavior of the process (e.g., time until the next arrival) is independent of the past history before time $t$.

-   **Time until next arrival**: The time until the next arrival, given that no arrival has occurred since some time $t_0$, is still Exponentially distributed with parameter $\lambda$. This is the **memorylessness property** of the Exponential distribution, which characterizes the inter-arrival times in a Poisson process.

-   **Inter-arrival Times ($T_k$)**: Let $Y_1, Y_2, \dots$ be the times of the 1st, 2nd, ..., arrivals respectively. The inter-arrival times are defined as $T_1 = Y_1$ (time until the first arrival) and $T_k = Y_k - Y_{k-1}$ for $k \ge 2$ (time between the $(k-1)$-th and $k$-th arrival).
    -   Each $T_k$ is an independent and identically distributed (i.i.d.) Exponential random variable with parameter $\lambda$.
    -   So, $E[T_k] = 1/\lambda$ and $Var(T_k) = 1/\lambda^2$ for all $k$.

-   **Time of the k-th arrival ($Y_k$) as a sum of i.i.d. exponentials**: Since $Y_k = T_1 + T_2 + \dots + T_k$, the time of the $k$-th arrival is the sum of $k$ i.i.d. Exponential random variables. This directly leads to $Y_k$ following an Erlang($k, \lambda$) distribution, as discussed previously.
    -   $E[Y_k] = E[T_1] + \dots + E[T_k] = k \cdot (1/\lambda) = k/\lambda$
    -   $Var(Y_k) = Var(T_1) + \dots + Var(T_k) = k \cdot (1/\lambda^2) = k/\lambda^2$

#### Plot: Number of Arrivals ($N_\tau$ - Poisson Distribution)

We simulate a Poisson process over a fixed time interval $\tau$ many times and count the number of arrivals in each simulation. The distribution of these counts should follow a Poisson distribution with parameter $\lambda\tau$.

````python
from scipy.stats import poisson
# Define a lambda rate for the Poisson process
lambda_rate = 2 # arrivals per unit time
def simulate_poisson_arrivals(lambda_rate, time_interval,num_simulations=10000):
    arrival_counts = []
    for _ in range(num_simulations):
        # Simulate a Poisson distribution for the number of arrivals
        num_arrivals = np.random.poisson(lambda_rate * time_interval)
        arrival_counts.append(num_arrivals)
    return np.array(arrival_counts)

# Simulate number of arrivals for a time interval of 5 units
time_interval_N = 5
poisson_arrivals_data = simulate_poisson_arrivals(lambda_rate, time_interval_N)

# Calculate theoretical mean and variance
theoretical_mean_N = lambda_rate * time_interval_N
theoretical_var_N = lambda_rate * time_interval_N

# Calculate empirical PMF from simulated data (for theoretical plot comparison)
unique_k, counts = np.unique(poisson_arrivals_data, return_counts=True)
empirical_pmf_values = counts / len(poisson_arrivals_data)

# Plotting
plt.figure(figsize=(13, 5), dpi=60)
# Create x values that span the range of observed data for theoretical PMF
k_vals_for_pmf = np.arange(0, np.max(poisson_arrivals_data) + 1)
theoretical_pmf = poisson.pmf(k_vals_for_pmf, theoretical_mean_N)
````

````python
from scipy.stats import expon
# Define a lambda rate for the Poisson process
lambda_rate = 2 # arrivals per unit time
def simulate_first_poisson_arrival_time(lambda_rate, num_simulations=10000):
    first_arrival_times = []
    for _ in range(num_simulations):
        # Time until first arrival in a Poisson process is Exp. distributed
        time = np.random.exponential(1/lambda_rate)
        first_arrival_times.append(time)
    return np.array(first_arrival_times)

first_arrival_data = simulate_first_poisson_arrival_time(lambda_rate)

# Calculate theoretical mean and variance for Exp(lambda)
theoretical_mean_exp = 1 / lambda_rate
theoretical_var_exp = 1 / (lambda_rate**2)

x_vals_exp = np.linspace(0, np.max(first_arrival_data), 500)
theoretical_pdf_exp = expon.pdf(x_vals_exp, scale=1/lambda_rate)
````

#### Plot: Time of the k-th Arrival ($Y_k$ - Erlang Distribution)

We simulate the time until the $k$-th arrival. This should follow an Erlang distribution with shape parameter $k$ and rate parameter $\lambda$.

````python
def simulate_kth_poisson_arrival_time(lambda_rate, k, num_simulations=10000):
    kth_arrival_times = []
    for _ in range(num_simulations):
        # Time until k-th arrival is a sum of k i.i.d. Exponential random variables
        # which follows an Erlang distribution
        time = np.sum(np.random.exponential(1/lambda_rate, k))
        kth_arrival_times.append(time)
    return np.array(kth_arrival_times)

# Let's simulate for k=3 successes to compare with its theoretical distribution
k_sim = 3
kth_arrival_data = simulate_kth_poisson_arrival_time(lambda_rate, k_sim)

# Define k values to plot theoretical Erlang PDFs
k_values_to_plot = [1, 2, 3]

# Determine max x-value for plotting to ensure all PDFs are visible
max_x_val = np.max(kth_arrival_data) * 1.5 # Extend the range a bit
x_vals_erlang = np.linspace(0, max_x_val, 500)

for i, k_val in enumerate(k_values_to_plot):
    theoretical_pdf_erlang = erlang.pdf(x_vals_erlang, k_val, scale=1/lambda_rate)
````

#### Plot: Inter-arrival Times ($T_k$ - i.i.d. Exponential Distribution)

We simulate a long Poisson process and collect all inter-arrival times. These times should be independent and identically distributed Exponential random variables.

````python
def simulate_poisson_inter_arrival_times(lambda_rate, total_time=10000):
    inter_arrival_times = []
    current_time = 0
    while current_time < total_time:
        next_inter_arrival = np.random.exponential(1/lambda_rate)
        current_time += next_inter_arrival
        if current_time < total_time:
            inter_arrival_times.append(next_inter_arrival)
    return np.array(inter_arrival_times)

inter_arrival_data_poisson = simulate_poisson_inter_arrival_times(lambda_rate)

# Calculate theoretical mean and variance for Exponential(lambda)
theoretical_mean_inter = 1 / lambda_rate
theoretical_var_inter = 1 / (lambda_rate**2)

x_vals_inter = np.linspace(0, np.max(inter_arrival_data_poisson), 500)
theoretical_pdf_inter = expon.pdf(x_vals_inter, scale=1/lambda_rate)
````

#### Merging Independent Poisson Processes

If we have two independent Poisson processes, $N_1(t)$ with rate $\lambda_1$ and $N_2(t)$ with rate $\lambda_2$, their merged process $N(t) = N_1(t) + N_2(t)$ is also a Poisson process with rate $\lambda = \lambda_1 + \lambda_2$.

Furthermore, if an arrival occurs in the merged process, the probability that it originated from process 1 is $\frac{\lambda_1}{\lambda_1 + \lambda_2}$, and the probability that it originated from process 2 is $\frac{\lambda_2}{\lambda_1 + \lambda_2}$. This is a direct application of Bayes' theorem and the properties of Poisson processes.

````python
def simulate_merged_poisson(lambda1, lambda2, time_interval, num_simulations=10000):
    merged_arrivals = 0
    first_process_given_merged = 0
    total_merged_events = 0

    for _ in range(num_simulations):
        # Simulate number of arrivals for each process in the given time interval
        arrivals1 = np.random.poisson(lambda1 * time_interval)
        arrivals2 = np.random.poisson(lambda2 * time_interval)

        # Total arrivals in the merged process for this simulation
        total_current_merged_arrivals = arrivals1 + arrivals2
        merged_arrivals += total_current_merged_arrivals

        # If there were any merged arrivals, count the contribution from process 1
        if total_current_merged_arrivals > 0:
            first_process_given_merged += arrivals1 # Summing up all process 1 arrivals
            total_merged_events += total_current_merged_arrivals

    # Empirical probability of an arrival in the merged process per unit time
    # This needs to be scaled by the time_interval to get a rate
    empirical_lambda_merged = merged_arrivals / (num_simulations * time_interval)

    # Empirical conditional probability
    empirical_conditional_p = first_process_given_merged / total_merged_events if total_merged_events > 0 else 0

    return empirical_lambda_merged, empirical_conditional_p

# Define rates for the two independent Poisson processes
lambda_process1 = 1.0  # arrivals per unit time
lambda_process2 = 1.5  # arrivals per unit time

time_interval_merged = 1.0 # Simulate over 1 unit of time to get rates directly

# Simulate the merged process
empirical_lambda_merged, empirical_conditional_p = simulate_merged_poisson(lambda_process1, lambda_process2, time_interval_merged)

# Theoretical probabilities
theoretical_lambda_merged = lambda_process1 + lambda_process2
theoretical_conditional_p = lambda_process1 / theoretical_lambda_merged
````

````python
def plot_merged_poisson_processes(lambda1, lambda2, time_interval=10.0):
    # Simulate arrivals for Process 1
    current_time_1 = 0
    arrivals_1 = []
    while current_time_1 < time_interval:
        inter_arrival = np.random.exponential(1/lambda1)
        current_time_1 += inter_arrival
        if current_time_1 < time_interval:
            arrivals_1.append(current_time_1)

    # Simulate arrivals for Process 2
    current_time_2 = 0
    arrivals_2 = []
    while current_time_2 < time_interval:
        inter_arrival = np.random.exponential(1/lambda2)
        current_time_2 += inter_arrival
        if current_time_2 < time_interval:
            arrivals_2.append(current_time_2)

    # Merge arrivals
    merged_arrivals = sorted(arrivals_1 + arrivals_2)

# Reuse the parameters defined earlier for lambda_process1 and lambda_process2
lambda_process1 = 1.0  # arrivals per unit time
lambda_process2 = 1.5  # arrivals per unit time
plot_merged_poisson_processes(lambda_process1, lambda_process2, time_interval=10.0)
````

#### Splitting of a Poisson Process

If a Poisson process $N(t)$ with rate $\lambda$ is split into two streams, $N_1(t)$ and $N_2(t)$, such that each arrival is independently directed to stream 1 with probability $p$ and to stream 2 with probability $1-p$, then
-   Both $N_1(t)$ and $N_2(t)$ are also Poisson processes.
-   The rate of $N_1(t)$ is $\lambda_1 = \lambda \cdot p$, and the rate of $N_2(t)$ is $\lambda_2 = \lambda \cdot (1-p)$.
-   Crucially, $N_1(t)$ and $N_2(t)$ are **independent** Poisson processes. This is a powerful property, unlike the splitting of Bernoulli processes where the resulting streams are not independent.

````python
def simulate_poisson_splitting(lambda_original, p_split, time_interval, num_simulations=10000):
    original_arrivals = 0
    stream1_arrivals = 0
    stream2_arrivals = 0

    for _ in range(num_simulations):
        # Simulate total arrivals in the original Poisson process
        total_current_arrivals = np.random.poisson(lambda_original * time_interval)
        original_arrivals += total_current_arrivals

        # Split each arrival independently
        for _ in range(total_current_arrivals):
            if random.random() < p_split:
                stream1_arrivals += 1
            else:
                stream2_arrivals += 1

    empirical_lambda_original = original_arrivals / (num_simulations * time_interval)
    empirical_lambda_stream1 = stream1_arrivals / (num_simulations * time_interval)
    empirical_lambda_stream2 = stream2_arrivals / (num_simulations * time_interval)

    return empirical_lambda_original, empirical_lambda_stream1, empirical_lambda_stream2

# Define parameters for the original process and splitting probability
lambda_original = 2.0  # arrivals per unit time
p_split = 0.4        # probability of an arrival going to stream 1
time_interval_split = 1.0

# Simulate the splitting
emp_lambda_orig, emp_lambda_s1, emp_lambda_s2 = simulate_poisson_splitting(lambda_original, p_split, time_interval_split)

# Theoretical probabilities
theoretical_lambda_s1 = lambda_original * p_split
theoretical_lambda_s2 = lambda_original * (1 - p_split)
````

#### Sum of two Independent Poisson Random Variables

If $X \sim \text{Poisson}(\lambda_1)$ and $Y \sim \text{Poisson}(\lambda_2)$ are two independent Poisson random variables, then their sum $Z = X + Y$ is also a Poisson random variable with parameter $\lambda = \lambda_1 + \lambda_2$.

This property is a direct consequence of the merging property of Poisson processes. If $X$ represents the number of arrivals in time $t$ from a Poisson process with rate $\lambda_1$, and $Y$ represents the number of arrivals in time $t$ from an independent Poisson process with rate $\lambda_2$, then $X+Y$ represents the total number of arrivals from the merged process, which we know is Poisson with rate $\lambda_1 + \lambda_2$ for the same time $t$. Therefore, the sum of the random variables follows a Poisson distribution with the sum of their parameters.

#### Sum of two Independent Erlang Random Variables

If $X \sim \text{Erlang}(k_1, \lambda)$ and $Y \sim \text{Erlang}(k_2, \lambda)$ are two independent Erlang random variables with the *same rate parameter* $\lambda$, then their sum $Z = X + Y$ is also an Erlang random variable with parameters $k = k_1 + k_2$ and $\lambda$.

This property stems from the fact that an Erlang($k, \lambda$) distribution is the sum of $k$ independent and identically distributed Exponential($\lambda$) random variables. Therefore, if $X$ is the sum of $k_1$ such exponentials and $Y$ is the sum of $k_2$ such exponentials, their independent sum $X+Y$ will be the sum of $k_1+k_2$ i.i.d. exponentials, which is an Erlang($k_1+k_2, \lambda$) distribution.

#### Poisson Properties and Edge Cases

**Dependence** is Key: The most crucial edge case is when the Poisson random variables are not independent. As discussed in the notebook under 'When the Sum of Two Poisson Random Variables is NOT Poisson', if $X$ and $Y$ are dependent (e.g., representing arrivals in overlapping intervals of the same Poisson process), their sum $X+Y$ will generally not be Poisson distributed.

**Homogeneity** (Constant Rate): The standard properties assume a homogeneous Poisson process, meaning the arrival rate ($\lambda$) is constant over time. If the rate changes over time (a non-homogeneous Poisson process), then calculations for the number of arrivals in an interval become more complex, and simple sum rules for $\lambda t$ might not apply directly without integration or segmenting the time interval.

Defining 'Doubling the Speed of Recording': It's worth clarifying what 'doubling the speed of recording' means. Usually, this implies the underlying event-generating process either genuinely doubles its rate, or we are effectively observing two independent, identical processes simultaneously. If, instead, it meant observing for twice the duration, then the parameter would change from $\lambda t$ to $\lambda (2t) = 2\lambda t$, which is mathematically equivalent to doubling $\lambda$ for the same unit time, but the conceptual origin is different.

Very Small Time Intervals or Rates: While the Poisson distribution handles all non-negative $\lambda t$, if $\lambda t$ approaches zero, the probability of zero events approaches 1, and the probabilities of 1 or more events become very small. This isn't a 'breakdown' of the model but a boundary of its practical relevance where events are exceedingly rare.

#### Probability Calculation for Poisson Process

Let $\lambda$ be the arrival rate per hour. We are interested in two time intervals: Interval 1: $I_1 = [10, 11)$, which has a duration of $\tau_1 = 1$ hour. & Interval 2: $I_2 = [10, 12)$, which has a duration of $\tau_2 = 2$ hours.

Let $N(t)$ denote the number of arrivals in a time interval of length $t$. For a Poisson process, $N(t)$ follows a Poisson distribution with parameter $\lambda t$. That is, $P(N(t) = k) = \frac{(\lambda t)^k e^{-\lambda t}}{k!}$.

Let $N_{[a,b]}$ denote the number of arrivals in the interval $[a,b]$. We want $P(N_{[10,11]} \le 1 \text{ and } N_{[10,12]} = 2)$. To solve this, we can split the interval $[10, 12)$ into two disjoint sub-intervals: $[10, 11)$ and $[11, 12)$.
Let $X = N_{[10,11]}$ be the number of arrivals between 10 and 11. $X \sim \text{Poisson}(\lambda \cdot 1) = \text{Poisson}(\lambda)$, and $Y = N_{[11,12]}$ be the number of arrivals between 11 and 12. $Y \sim \text{Poisson}(\lambda \cdot 1) = \text{Poisson}(\lambda)$.

Since these intervals are disjoint, $X$ and $Y$ are independent random variables. The condition $N_{[10,12]} = 2$ can be expressed as $X + Y = 2$. The condition $N_{[10,11]} \le 1$ can be expressed as $X \le 1$. To calculate $P(X \le 1 \text{ and } X + Y = 2)$, this can be broken down into two mutually exclusive cases:

Let's define the probabilities for a single unit of time (e.g., 1 hour):
*   $a = P(N_1=0) = \frac{(\lambda \cdot 1)^0 e^{-\lambda \cdot 1}}{0!} = e^{-\lambda}$
*   $b = P(N_1=1) = \frac{(\lambda \cdot 1)^1 e^{-\lambda \cdot 1}}{1!} = \lambda e^{-\lambda}$
*   $c = P(N_1=2) = \frac{(\lambda \cdot 1)^2 e^{-\lambda \cdot 1}}{2!} = \frac{\lambda^2 e^{-\lambda}}{2}$

Case 1: $X = 0$ and $X + Y = 2$

This implies $X = 0$ and $Y = 2$.
$P(X=0 \text{ and } Y=2) = P(X=0) \cdot P(Y=2)$ (due to independence).
Using our definitions:
$P(X=0) = a$
$P(Y=2) = c$
$P(\text{Case 1}) = a \cdot c$

Case 2: $X = 1$ and $X + Y = 2$

This implies $X = 1$ and $Y = 1$.
$P(X=1 \text{ and } Y=1) = P(X=1) \cdot P(Y=1)$ (due to independence).
Using our definitions:
$P(X=1) = b$
$P(Y=1) = b$
$P(\text{Case 2}) = b \cdot b = b^2$

Total Probability

The total probability is the sum of the probabilities of these two disjoint cases:
$P(\text{total}) = P(\text{Case 1}) + P(\text{Case 2})$ .
$P(\text{total}) = a \cdot c + b^2$

Substituting back the original expressions for $a, b, c$:
$P(\text{total}) = (e^{-\lambda}) \left( \frac{\lambda^2 e^{-\lambda}}{2} \right) + (\lambda e^{-\lambda})^2$
$P(\text{total}) = \frac{\lambda^2 e^{-2\lambda}}{2} + \lambda^2 e^{-2\lambda}$
$P(\text{total}) = \lambda^2 e^{-2\lambda} \left( \frac{1}{2} + 1 \right)$
$P(\text{total}) = \frac{3}{2} \lambda^2 e^{-2\lambda}$

Therefore, the probability that you have at most one arrival between 10-11 and exactly two arrivals between 10-12 is $\frac{3}{2} \lambda^2 e^{-2\lambda}$.

#### Problem: Probability of exactly two arrivals for each type in a specified time interval

Given three independent Poisson processes for items of type A, B, and C with arrival rates $\lambda_A$, $\lambda_B$, and $\lambda_C$ respectively. Let $T = 5$ minutes be the time interval.

Let $X$ be the event that exactly two items of type A arrive in 5 minutes.
Let $Y$ be the event that exactly two items of type B arrive in 5 minutes.
Let $Z$ be the event that exactly two items of type C arrive in 5 minutes.

Since the processes are independent, $X, Y, Z$ are independent events.
We want to calculate the probability $P(X \text{ and } Y \text{ and } Z) = P(X) \cdot P(Y) \cdot P(Z)$.

For a Poisson process with rate $\lambda$ and time interval $\tau$, the probability of exactly $k$ arrivals is given by $P(N_\tau = k) = \frac{(\lambda\tau)^k e^{-\lambda\tau}}{k!}$.

Thus, $P(X) = P(N_A(5) = 2) = \frac{(\lambda_A \cdot 5)^2 e^{-(\lambda_A \cdot 5)}}{2!}$
$P(Y) = P(N_B(5) = 2) = \frac{(\lambda_B \cdot 5)^2 e^{-(\lambda_B \cdot 5)}}{2!}$
$P(Z) = P(N_C(5) = 2) = \frac{(\lambda_C \cdot 5)^2 e^{-(\lambda_C \cdot 5)}}{2!}$

The overall probability is the product: $P(X)P(Y)P(Z) = \left( \frac{(5\lambda_A)^2 e^{-5\lambda_A}}{2} \right) \left( \frac{(5\lambda_B)^2 e^{-5\lambda_B}}{2} \right) \left( \frac{(5\lambda_C)^2 e^{-5\lambda_C}}{2} \right)$

#### When the Sum of Two Poisson Random Variables is NOT Poisson (Dependent RVs)

The property that the sum of two Poisson random variables is also a Poisson random variable only holds if the two random variables are independent. If the two Poisson random variables are dependent, their sum is generally not Poisson.

Let's consider a use case involving overlapping time intervals from the *same* Poisson process.

**Scenario:** Counting arrivals in overlapping intervals of a single Poisson Process.

Suppose we have a single Poisson process $N(t)$ with a constant arrival rate $\lambda$. Let's define two random variables
*   $X = N(t_1)$: The number of arrivals in the time interval $[0, t_1]$. $X \sim \text{Poisson}(\lambda t_1)$.
*   $Y = N(t_2)$: The number of arrivals in the time interval $[0, t_2]$. $Y \sim \text{Poisson}(\lambda t_2)$.

Now, let's consider the case where the two intervals *overlap*. For example, assume $0 < t_1 < t_2$.

In this situation, $X$ and $Y$ are clearly **dependent**. The arrivals counted in $X$ (up to time $t_1$) are a subset of the arrivals counted in $Y$ (up to time $t_2$). Specifically, $Y$ can be written as $Y = N(t_1) + N((t_1, t_2])$, or $Y = X + N_{additional}$, where $N_{additional}$ is the number of arrivals in the interval $(t_1, t_2]$ and $N_{additional} \sim \text{Poisson}(\lambda(t_2-t_1))$. Importantly, $N_{additional}$ is independent of $X$.

Now, let's look at the sum $Z = X + Y$. Substituting $Y = X + N_{additional}$:
$Z = X + (X + N_{additional}) = 2X + N_{additional}$.

To check if $Z$ is Poisson, we need to compare its mean and variance. For a Poisson random variable, the mean and variance must be equal.

*   **Mean of Z:** $E[Z] = E[2X + N_{additional}] = 2E[X] + E[N_{additional}]$. $E[Z] = 2(\lambda t_1) + (\lambda (t_2 - t_1))$. $E[Z] = 2\lambda t_1 + \lambda t_2 - \lambda t_1 = \lambda t_1 + \lambda t_2$

*   **Variance of Z:** (Since $X$ and $N_{additional}$ are independent). $Var[Z] = Var[2X + N_{additional}] = Var[2X] + Var[N_{additional}]$. $Var[Z] = (2^2)Var[X] + Var[N_{additional}]$. $Var[Z] = 4(\lambda t_1) + (\lambda (t_2 - t_1))$. $Var[Z] = 4\lambda t_1 + \lambda t_2 - \lambda t_1 = 3\lambda t_1 + \lambda t_2$

For $Z$ to be a Poisson rv, we would need $E[Z] = Var[Z]$:
$\lambda t_1 + \lambda t_2 = 3\lambda t_1 + \lambda t_2$
$\lambda t_1 = 3\lambda t_1$ .

This equality only holds if $2\lambda t_1 = 0$, which implies either $\lambda = 0$ (no arrivals) or $t_1 = 0$ (no initial interval). In any non-trivial scenario where $\lambda > 0$ and $t_1 > 0$, we have $E[Z] \ne Var[Z]$.

**Conclusion:** Because $E[Z] \ne Var[Z]$ (for $\lambda > 0, t_1 > 0$), the sum $Z = X+Y$ is not a Poisson random variable when $X$ and $Y$ are dependent (e.g., from overlapping intervals of the same Poisson process).

#### Compound Distributions: Sum of a Binomial and a Poisson Distributed Number of Random Variables

Let $X_1, X_2, \dots, X_n$ be independent Bernoulli random variables with parameter $p$. Let $N$ be a random variable that takes integer values and is independent of $X_i$ for $i = 1, 2, \dots$. Let $Y = \sum_{i=1}^{N} X_i$ for positive values of $N$, and let $Y = 0$ when $N = 0$. This describes a **compound distribution**, where the number of terms in the sum (the $N$ Bernoulli random variables) is itself a random variable.

**1) If $N$ is Binomial with parameters $m$ and $q$, then $Y$ is Binomial with parameters $m$ and $p \cdot q$.**

*   **Explanation:** This is a classic result for compound binomial distributions. When you have a sum of $N$ independent Bernoulli trials (each with success probability $p$), where $N$ itself follows a Binomial distribution ($m$ trials, $q$ success probability), the resulting sum $Y$ also follows a Binomial distribution. The number of trials for $Y$ remains $m$, and the new success probability becomes the product of the individual Bernoulli success probability $p$ and the success probability of $N$, which is $q$. This can be formally proven using Probability Generating Functions (PGFs).

*   **Intuition:** Imagine $m$ machines, each having a $q$ probability of being active. If a machine is active, it performs one Bernoulli trial with probability $p$ of success. If it's not active, it produces 0 successes. So, for each machine, the probability of it producing a 'success' (being active AND performing a successful Bernoulli trial) is $q \cdot p$. Since there are $m$ such independent machines, the total number of successes $Y$ will follow a Binomial distribution with parameters $m$ and $qp$.

**2) If $N$ is Poisson with parameter $\lambda$, then $Y$ is Poisson with parameter $\lambda \cdot p$.**

*   **Explanation:** This statement is also **correct**. This is a fundamental result for compound Poisson distributions. If $Y$ is a sum of $N$ independent Bernoulli trials (each with success probability $p$), and $N$ follows a Poisson distribution with parameter $\lambda$, then $Y$ itself follows a Poisson distribution with a new parameter $\lambda p$. Similar to the first case, this is elegantly proven using Probability Generating Functions.

*   **Intuition:** Consider a Poisson process where events occur at an average rate of $\lambda$. If each event, independently, has a probability $p$ of being a 'type A' event and $1-p$ of being a 'type B' event, then the number of 'type A' events (which is $Y$ in this context) will also form a Poisson process with an average rate of $\lambda p$. This is often called **Poisson thinning** or **Poisson splitting**.

These properties are incredibly useful in modeling real-world phenomena where the number of occurrences of an event is random, and each occurrence then has an independent probability of leading to a 'success' or specific outcome.

#### Compound Distributions: Sum of a Geometrically Distributed Number of Random Variables

Let $N$ be a Geometric random variable with parameter $q$ (representing the number of trials until the first success, including the success, so $P(N=k) = (1-q)^{k-1}q$ for $k=1, 2, \dots$).
Let $X_1, X_2, \dots$ be i.i.d. random variables, and also independent of $N$.
Let $Y = \sum_{i=1}^{N} X_i$ be the sum of these random variables, where the number of terms in the sum is random and determined by $N$.

**1) If $X_i$ is Geometric with parameter $p$, then $Y$ is Geometric with parameter $pq$.**

*   **Explanation:** This is a well-known result in probability theory for compound geometric distributions. It can be formally proven using Probability Generating Functions (PGFs).
    *   The PGF of a Geometric$(p)$ distribution is $G_X(z) = \frac{pz}{1-(1-p)z}$.
    *   The PGF of $N \sim \text{Geometric}(q)$ is $G_N(z) = \frac{qz}{1-(1-q)z}$.
    *   The PGF of the compound sum $Y$ is given by $G_Y(z) = G_N(G_X(z))$.
    *   Substituting $G_X(z)$ into $G_N(z)$:$G_Y(z) = \frac{q \left(\frac{pz}{1-(1-p)z}\right)}{1-(1-q)\left(\frac{pz}{1-(1-p)z}\right)} = \frac{qpz}{1-(1-p)z - (1-q)pz} = \frac{qpz}{1 - z(1-qp)}$
    *   This is precisely the PGF of a Geometric distribution with parameter $qp$.

*   **Intuition:** Imagine a two-stage process. First, you're waiting for a "major event", which happens with probability $q$ at each trial. If it doesn't happen, you try again. This waiting time is $N \sim \text{Geometric}(q)$. Now, for *each* of those trials you are performing (the $N$ trials), you are also waiting for another "sub-event" (an $X_i$ success) with probability $p$. The compound distribution $Y$ effectively models the number of sub-events that happen until the major event occurs. Since each major trial (until $N$) can be thought of as having a chance $q$ of success, and each of those successful major trials then has a chance $p$ of a sub-event success, the overall effective success probability for the compound event is $qp$. Think of it as a hierarchical sequence of Bernoulli trials.

**2) If $X_i$ is Exponential with parameter $\lambda$, then $Y$ is Exponential with parameter $\lambda q$.**

*   **Explanation:** This is a powerful result often encountered in continuous-time Markov chains and queuing theory. It can be formally proven using Moment Generating Functions (MGFs) or Laplace Transforms.
    *   The MGF of an Exponential($\lambda$) distribution is $M_X(s) = \frac{\lambda}{\lambda - s}$.
    *   The PGF of $N \sim \text{Geometric}(q)$ (as defined before, number of trials) is $G_N(z) = \frac{qz}{1-(1-q)z}$.
    *   The MGF of the compound sum $Y$ is given by $M_Y(s) = G_N(M_X(s))$.
    *   Substituting $M_X(s)$ into $G_N(z)$ (where $z$ is replaced by $M_X(s)$):$M_Y(s) = \frac{q \left(\frac{\lambda}{\lambda - s}\right)}{1-(1-q)\left(\frac{\lambda}{\lambda - s}\right)} = \frac{q\lambda}{\lambda - s - (1-q)\lambda} = \frac{q\lambda}{\lambda - s - \lambda + q\lambda} = \frac{q\lambda}{q\lambda - s}$
    *   This is precisely the MGF of an Exponential distribution with parameter $\lambda q$.

*   **Intuition:** Consider a process where events arrive according to an Exponential distribution (e.g., $X_i$ are inter-arrival times). Now, imagine you only count these events if a "gate" is open, and the gate closes after a geometrically distributed number of trials ($N$) fail to open it. Or, more simply, think about the probability of success $q$ for the geometric variable. If $X_i$ represents the duration of an event, and you have $N$ such events, where $N$ itself is geometric with success probability $q$, it implies that you are essentially scaling the rate at which these exponential durations accumulate. A geometric distribution describes waiting for a 'success' in a sequence of Bernoulli trials. If the $X_i$ are times, and you are waiting for $N$ of them, and $N$ depends on a success probability $q$, then the combined process effectively 'speeds up' or 'slows down' the rate. For a geometric with success probability $q$ (number of trials until first success), if we have an exponential process with rate $\lambda$, the compound sum will have an effective rate of $\lambda q$. This is similar to how splitting a Poisson process (where inter-arrival times are exponential) into a substream results in a lower rate.

#### Properties of the k-th Arrival Time ($Y_k$) in a Poisson Process

Let $Y_k$ be the time of the $k$-th arrival in a Poisson process with rate parameter $\lambda = 1$.

1.  **Expectation of $Y_k$:** The expectation of $Y_k$ is $E[Y_k] = k/\lambda$. Given $\lambda = 1$, then $E[Y_k] = k$.

2.  **Probability $P(Y_k \ge k)$:** The statement $P(Y_k \ge k) = 1/2$ for any finite $k$ is generally **incorrect**. For example, if $k=1$ and $\lambda=1$, $P(Y_1 \ge 1) = P(N(1) = 0) = e^{-\lambda \cdot 1} = e^{-1} \approx 0.3678$, which is not $1/2$.

3.  **Limit of $P(Y_k \ge k)$ as $k \to \infty$:** The statement that $\lim_{k\to\infty} P(Y_k \ge k) = 1/2$ is **correct**. This property can be understood through the Central Limit Theorem. The random variable $Y_k$ is the sum of $k$ independent and identically distributed Exponential random variables (inter-arrival times), each with mean $1/\lambda$ and variance $1/\lambda^2$.
    
    For large $k$, $Y_k$ can be approximated by a Normal distribution with:
    *   Mean: $E[Y_k] = k/\lambda = k$ (since $\lambda = 1$)
    *   Variance: $Var(Y_k) = k/\lambda^2 = k$ (since $\lambda = 1$)
    
    Therefore, for large $k$:
    $P(Y_k \ge k) = P\left(\frac{Y_k - E[Y_k]}{\sqrt{Var(Y_k)}} \ge \frac{k - k}{\sqrt{k}}\right) = P(Z \ge 0)$
    Where $Z$ is a standard normal random variable. The probability $P(Z \ge 0)$ is indeed $1/2$. Thus, as $k$ approaches infinity, the probability $P(Y_k \ge k)$ converges to $1/2$.

#### Classic Problem: Time Until the First (or Last) Lightbulb Burns Out

Consider a scenario where you have $n$ identical lightbulbs, and the lifetime of each lightbulb is an independent and identically distributed (i.i.d.) Exponential random variable with rate parameter $\lambda$. We are interested in two main questions:

1.**Time until the First Lightbulb Burns Out ($T_{\text{min}}$):** This is the minimum of $n$ i.i.d. Exponential random variables.
If $X_1, X_2, \dots, X_n$ are i.i.d. $\text{Exponential}(\lambda)$ random variables, then $T_{\text{min}} = \min(X_1, X_2, \dots, X_n)$ is also an Exponential random variable, but with rate parameter $n\lambda$.
    
*   **Derivation (Intuitive):** The probability that any single bulb is still working at time $t$ is $P(X_i > t) = e^{-\lambda t}$. For *all* $n$ bulbs to still be working at time $t$, given their independence, the probability is $P(T_{\text{min}} > t) = P(X_1 > t, \dots, X_n > t) = \prod_{i=1}^n P(X_i > t) = (e^{-\lambda t})^n = e^{-n\lambda t}$. This is the survival function of an $\text{Exponential}(n\lambda)$ distribution.

2.**Time until the Last Lightbulb Burns Out ($T_{\text{max}}$):** This is the maximum of $n$ i.i.d. Exponential random variables.
If $X_1, X_2, \dots, X_n$ are i.i.d. $\text{Exponential}(\lambda)$ random variables, then $T_{\text{max}} = \max(X_1, X_2, \dots, X_n)$ does *not* follow a simple Exponential distribution. Its distribution is more complex.
    
*   **Derivation (Intuitive):** The probability that all bulbs have burned out by time $t$ is $P(T_{\text{max}} \le t) = P(X_1 \le t, \dots, X_n \le t) = \prod_{i=1}^n P(X_i \le t) = (1 - e^{-\lambda t})^n$. This is the CDF of $T_{\text{max}}$.
*   The PDF can be found by differentiating the CDF: $f_{T_{\text{max}}}(t) = n\lambda e^{-\lambda t} (1 - e^{-\lambda t})^{n-1}$.
*   The expectation is $E[T_{\text{max}}] = \sum_{k=1}^n \frac{1}{k\lambda} = \frac{1}{\lambda} \sum_{k=1}^n \frac{1}{k}$. This is related to the $n$-th harmonic number.

#### Pictorial View: Time until the First (or Last) Lightbulb Burns Out

Let's consider the scenario with three lightbulbs:
-   Independent lifetimes $X, Y, Z \sim \text{exp}(\lambda)$

The expected time until the first burnout ($T_{\text{min}}$) is $E[T_{\text{min}}] = E[\min\{X, Y, Z\}] = \frac{1}{3\lambda}$.

This can be derived by considering the probability that $\min\{X, Y, Z\} \ge t$:
$P(\min\{X, Y, Z\} \ge t) = P(X \ge t, Y \ge t, Z \ge t)$
Due to independence:
$P(X \ge t, Y \ge t, Z \ge t) = P(X \ge t) \cdot P(Y \ge t) \cdot P(Z \ge t)$
$= e^{-\lambda t} \cdot e^{-\lambda t} \cdot e^{-\lambda t} = e^{-3\lambda t}$

This is the survival function of an Exponential distribution with rate $3\lambda$. Thus, $T_{\text{min}} \sim \text{exp}(3\lambda)$, and its expectation is $E[T_{\text{min}}] = \frac{1}{3\lambda}$.

This aligns with the concept that if $X, Y, Z$ are first arrivals in independent Poisson processes, then the minimum of these times (the first overall arrival) corresponds to the first arrival in a merged Poisson process with rate $3\lambda$.

The expectation for the time until the last burnout ($T_{\text{max}}$) for three bulbs is $E[T_{\text{max}}] = \frac{1}{\lambda} + \frac{1}{2\lambda} + \frac{1}{3\lambda}$. This is generally given by $E[T_{\text{max}}] = \frac{1}{\lambda} \sum_{k=1}^n \frac{1}{k}$ for $n$ bulbs. $Var(T_{\text{max}}) = \frac{1}{\lambda^2} \sum_{k=1}^n \frac{1}{k^2}$

#### Poisson Random Incidence (or PASTA: Poisson Arrivals See Time Averages)

Poisson Random Incidence, often referred to as the **PASTA (Poisson Arrivals See Time Averages) property**, is a fundamental concept in the study of Poisson processes and queuing theory. It states that for a system observed by a Poisson process, the proportion of arrivals that find the system in a particular state is the same as the proportion of time the system spends in that state.

In simpler terms, if events arrive according to a Poisson process, an arriving event sees the system "as if it arrived at a random time." This means that an arrival isn't more likely to occur during periods when the system is in a specific state (e.g., busy or idle), but rather observes the system's states in proportion to their long-run time averages.

**Formal Statement (Intuitive):**
If a system's state evolves over time, and its state is observed by a Poisson arrival process, then the probability that an arriving customer finds the system in a certain state $S$ is equal to the long-run proportion of time the system spends in state $S$.

$P(\text{system in state } S \mid \text{an arrival occurs}) = P(\text{system in state } S \text{ at an arbitrary time})$

**Key Implications:**
*   **Independence:** The property stems from the memorylessness of the Poisson process. The occurrence of an arrival is independent of the system's past and current state. Therefore, an arriving customer does not preferentially select certain system states.
*   **Simplification of Analysis:** PASTA simplifies the analysis of many stochastic systems, especially in queuing models. It allows us to relate time-average probabilities (e.g., the long-run proportion of time a server is busy) to arrival-average probabilities (e.g., the probability that an arriving customer finds the server busy).
*   **Common Use Case:** Consider a queuing system where customers arrive according to a Poisson process. The PASTA property implies that the fraction of arriving customers who have to wait is equal to the fraction of time the server is busy, assuming the system has reached a steady state.

**Contrast with Non-Poisson Arrivals:**
If arrivals are not Poisson, the PASTA property generally does not hold. For instance, if arrivals tend to occur in bursts (e.g., a deterministic arrival schedule or a highly clustered arrival process), an arriving customer might be more likely to find the system in a certain state (e.g., busy if many customers arrive together), and thus, what an arriving customer "sees" would not be representative of the time-average behavior.

#### Expected Length of the Interarrival Interval Containing an Observation

When an observer arrives at a random time $t^*$ to a Poisson process, they will land within some interarrival interval $(U, V)$, where $U$ is the time of the last arrival before $t^*$, and $V$ is the time of the next arrival after $t^*$.

Despite the individual interarrival times being Exponentially distributed with mean $1/\lambda$, the interarrival interval that a random observer lands in actually has a larger expected length. This is a consequence of **length-biasing**; longer intervals are more likely to contain the random observation point.

The length of the interarrival interval $S = V - U$ can be expressed as the sum of the time from the last arrival to the observation point and the time from the observation point to the next arrival:

$S = (t^* - U) + (V - t^*)$

Due to the memoryless property of the Poisson process and the fact that $t^*$ is a random observation point, both $(t^* - U)$ and $(V - t^*)$ are themselves independent Exponential random variables, each with rate $\lambda$. (Note: This is an intuitive explanation for the memoryless property extending to both 'backward' and 'forward' recurrence times).

Therefore, the expected length of this interarrival interval is:

$E[S] = E[(t^* - U)] + E[(V - t^*)] = \frac{1}{\lambda} + \frac{1}{\lambda} = \frac{2}{\lambda}$

And the variance of this interarrival interval is:

$Var(S) = Var(t^* - U) + Var(V - t^*) = \frac{1}{\lambda^2} + \frac{1}{\lambda^2} = \frac{2}{\lambda^2}$

This result means that an interarrival interval containing a random observation point is, on average, twice as long as an 'ordinary' interarrival interval (which has an expected length of $1/\lambda$).

#### Pictorial View: Random Incidence and Interarrival Interval

When an observer arrives at a random time $t^*$ within a Poisson process, they land in an interarrival interval defined by the time of the last arrival $U$ and the time of the next arrival $V$.

```
-------U-------t*---------V------> Time Axis
       <--------------S = V - U---------------->
       <---(t* - U)---><---(V - t*)--->
```

Where:
-   $U$: Time of the last arrival before $t^*$.
-   $V$: Time of the next arrival after $t^*$.
-   $t^*$: The random time the observer arrives.
-   $S = V - U$: The total length of the interarrival interval that contains $t^*$.
-   $(t^* - U)$: The time from the last arrival until the observer's arrival (backward recurrence time).
-   $(V - t^*)$: The time from the observer's arrival until the next arrival (forward recurrence time).

As discussed, due to the memoryless property of the Exponential distribution (which governs interarrival times in a Poisson process), both $(t^* - U)$ and $(V - t^*)$ are themselves independent Exponential random variables with rate $\lambda$. Therefore, the expected length of the interval $S$ is $E[S] = E[t^* - U] + E[V - t^*] = 1/\lambda + 1/\lambda = 2/\lambda$.

````python
import numpy as np

def simulate_interarrival_interval_variance(lambda_rate, total_process_duration, num_observations=10000):
    inter_arrival_times_S = []

    for _ in range(num_observations):
        # Simulate a long Poisson process to have enough arrival times
        arrival_times = []
        current_time = 0
        while current_time < total_process_duration:
            inter_arrival = np.random.exponential(1/lambda_rate)
            current_time += inter_arrival
            if current_time < total_process_duration:
                arrival_times.append(current_time)

        if not arrival_times: # If no arrivals in the duration, skip this observation
            continue

        # Choose a random observation time t* within the total process duration
        t_star = np.random.uniform(0, total_process_duration)

        # Find the last arrival (U) before t_star and the next arrival (V) after t_star
        U = 0.0 # Default if t_star is before the first arrival
        V = total_process_duration # Default if t_star is after the last arrival

        # Find U
        arrivals_before_t_star = [t for t in arrival_times if t <= t_star]
        if arrivals_before_t_star:
            U = max(arrivals_before_t_star)

        # Find V
        arrivals_after_t_star = [t for t in arrival_times if t > t_star]
        if arrivals_after_t_star:
            V = min(arrivals_after_t_star)
        else: # If t_star is after all simulated arrivals, V is effectively infinity (or end of duration)
            continue # Skip this observation as we can't determine S accurately

        # Calculate S = V - U
        inter_arrival_times_S.append(V - U)

    inter_arrival_times_S = np.array(inter_arrival_times_S)

    # Calculate empirical mean and variance
    empirical_mean_S = np.mean(inter_arrival_times_S)
    empirical_var_S = np.var(inter_arrival_times_S)

    # Calculate theoretical mean and variance
    theoretical_mean_S = 2 / lambda_rate
    theoretical_var_S = 2 / (lambda_rate**2)

# Set parameters for the simulation
lambda_rate_sim = 2.0 # events per unit time
total_process_duration_sim = 10000 # A long enough duration for the underlying Poisson process
num_observations_sim = 10000 # Number of times to pick a random t* and measure S

simulate_interarrival_interval_variance(lambda_rate_sim, total_process_duration_sim, num_observations_sim)
````

#### Visualize PASTA: Simulating Poisson Arrivals Seeing Time Averages

This simulation demonstrates the PASTA property. We will model a simple system that alternates between being in an 'idle' state and a 'busy' state. We then generate Poisson arrivals and check the state of the system at each arrival time. According to PASTA, the proportion of arrivals that find the system in a 'busy' state should be approximately equal to the long-run proportion of time the system spends in the 'busy' state.

````python
import numpy as np
import matplotlib.pyplot as plt

def simulate_system_state(duration, t_busy, t_idle):
    """Simulates a system alternating between busy and idle states."""
    system_states = []
    current_time = 0.0
    state = 'idle' # Start idle

    while current_time < duration:
        if state == 'idle':
            end_time = min(current_time + t_idle, duration)
            system_states.append((current_time, end_time, 'idle'))
            current_time = end_time
            state = 'busy'
        else: # state == 'busy'
            end_time = min(current_time + t_busy, duration)
            system_states.append((current_time, end_time, 'busy'))
            current_time = end_time
            state = 'idle'
    return system_states

def get_system_state_at_time(system_states, query_time):
    """Determines the system's state at a specific query_time."""
    for start, end, state in system_states:
        if start <= query_time < end:
            return state
    return None # Should not happen if query_time is within simulated duration
````

````python
def simulate_pasta(lambda_arrivals, t_busy, t_idle, simulation_duration, num_simulations=1000):
    arrival_observations = []
    total_time_busy = 0.0

    for _ in range(num_simulations):
        # Simulate system state for a long duration
        system_states = simulate_system_state(simulation_duration, t_busy, t_idle)

        # Generate Poisson arrivals over the same duration
        # The number of arrivals in time_interval follows Poisson(lambda * time_interval)
        num_poisson_arrivals = np.random.poisson(lambda_arrivals * simulation_duration)
        poisson_arrival_times = np.sort(np.random.uniform(0, simulation_duration, num_poisson_arrivals))

        busy_arrivals = 0
        for arrival_time in poisson_arrival_times:
            state_at_arrival = get_system_state_at_time(system_states, arrival_time)
            if state_at_arrival == 'busy':
                busy_arrivals += 1

        if num_poisson_arrivals > 0:
            arrival_observations.append(busy_arrivals / num_poisson_arrivals)
        else:
            # If no arrivals, we can't observe anything for this simulation run
            pass

    # Calculate theoretical time average busy proportion
    theoretical_time_avg_busy = t_busy / (t_busy + t_idle)

    # Calculate empirical arrival average busy proportion
    empirical_arrival_avg_busy = np.mean(arrival_observations) if arrival_observations else 0.0
````

````python
# Parameters for the simulation
lambda_arrivals = 5.0          # Average Poisson arrival rate (e.g., 5 arrivals per unit time)
t_busy = 3.0                   # Duration system is busy
t_idle = 2.0                   # Duration system is idle
simulation_duration = 100.0    # Duration for each individual simulation run
num_simulations = 1000         # Number of simulation runs to average over

# Run the PASTA simulation
simulate_pasta(lambda_arrivals, t_busy, t_idle, simulation_duration, num_simulations)
````

#### Non-Poisson Random Incidence (When PASTA Doesn't Hold)

The PASTA (Poisson Arrivals See Time Averages) property is a powerful and convenient characteristic that relies heavily on the **memorylessness** of the Poisson arrival process. When the arrival process is *not* Poisson, the PASTA property generally **does not hold**.

This means that for non-Poisson arrivals, the proportion of arrivals that find the system in a particular state is *not* necessarily the same as the proportion of time the system spends in that state.

**Why PASTA Fails for Non-Poisson Arrivals:**

1.**Lack of Memorylessness:** Non-Poisson processes are not memoryless. The probability of an arrival in the next instant *depends on the history* of previous arrivals. This means that an arrival event is not independent of the system's state leading up to that arrival.

2.**Arrivals Influence System State:** If arrivals tend to be clustered (e.g., bursty traffic) or deterministic (e.g., scheduled arrivals), an arriving entity might be more likely to find the system in a state caused by preceding arrivals. For example:
*   **Clustered Arrivals:** If arrivals come in bursts, the first arrival in a burst might find the system idle, but subsequent arrivals in the same burst are much more likely to find the system busy (or in a state of high activity) because of the recent influx.
*   **Deterministic Arrivals:** If customers arrive precisely every 10 minutes, and the service time is consistently 8 minutes, an arriving customer will *always* find the server idle for the first 2 minutes of a 10-minute cycle and busy for the remaining 8 minutes. The probability of finding the server busy upon arrival would be 0.8, which would correspond to the time-average if the system always followed this pattern. However, if service times vary, or if the arrival pattern is not completely deterministic, the time-average and arrival-average can diverge.

3.**Bias in Observation:** Non-Poisson arrivals effectively 'bias' their own observations of the system. They are more likely to arrive during periods when the system state is more conducive to their arrival pattern. This contradicts the 'random observer' nature implied by Poisson arrivals.

**Example:**
Consider a server that processes tasks. If tasks arrive in a non-Poisson manner (e.g., in batches), the probability that an *arriving task* finds the server busy might be higher than the overall time-average probability that the server is busy. This is because tasks arriving in a batch are likely to find the server already dealing with tasks from the same batch that arrived just before them. In contrast, a Poisson arrival is equally likely to hit any point in time, seeing the system as a truly random observer would.

In essence, for non-Poisson processes, the arrival process itself dictates how the system is observed, breaking the crucial independence required for PASTA.

#### Random Incidence in Non-Poisson Processes (Renewal Processes)

In contrast to Poisson processes where interarrival times are Exponentially distributed, in a **renewal process**, the interarrival times are independent and identically distributed (i.i.d.) according to some general distribution $F(x)$. When an observer arrives at a random time in such a process, the interval they land in (the *sampled interval*) is still subject to **length-biased sampling** – longer intervals are more likely to contain the random observation point.

Let $S_i$ be the length of the $i$-th interarrival interval. If an observer arrives at a random time, the probability that they arrive during an interval of length $x$ is proportional to $x$. More precisely, if the interarrival times have a probability density function (PDF) $f(x)$ and mean $E[S]$, the PDF of the length of the interval containing a random observation, $S^*$, is given by $f_{S^*}(x) = \frac{x}{E[S]} f(x)$.

The expected length of the observed interval, $E[S^*]$, is then $E[S^*] = \int_0^{\infty} x f_{S^*}(x) dx = \int_0^{\infty} x \frac{x}{E[S]} f(x) dx = \frac{E[S^2]}{E[S]}$.

For discrete interarrival times, if the interarrival times can take values $x_j$ with probability $P(S=x_j)$, then $E[S] = \sum_j x_j P(S=x_j)$ and $E[S^2] = \sum_j x_j^2 P(S=x_j)$. The probability that a random observer lands in an interval of length $x_j$ is then proportional to $x_j$, specifically $P(S^*=x_j) = \frac{x_j P(S=x_j)}{E[S]}$.


#### Problem: Discrete Interarrival Times

Consider a renewal process where the consecutive interarrival times are i.i.d. random variables, equally likely to be 5, 10, or 15 minutes. This means:
- $P(S = 5 	ext{ min}) = 1/3$
- $P(S = 10 	ext{ min}) = 1/3$
- $P(S = 15 	ext{ min}) = 1/3$

First, let's calculate the mean $E[S]$ and second moment $E[S^2]$ of the interarrival times:

$E[S] = (5 \cdot 1/3) + (10 \cdot 1/3) + (15 \cdot 1/3) = (5+10+15)/3 = 30/3 = 10$ minutes.

$E[S^2] = (5^2 \cdot 1/3) + (10^2 \cdot 1/3) + (15^2 \cdot 1/3) = (25 + 100 + 225)/3 = 350/3 \approx 116.67$ min$^2$.

Now, if an observer arrives at a random time, we want to find the expected length of the interarrival interval they land in, $E[S^*]$. Using the length-biased formula:

$E[S^*] = \frac{E[S^2]}{E[S]} = \frac{350/3}{10} = \frac{350}{30} = \frac{35}{3} \approx 11.67$ minutes.

Notice that $E[S^*] \approx 11.67$ minutes, which is greater than the average interarrival time of $E[S] = 10$ minutes. This confirms the length-biasing effect.

Alternatively, we can first find the probabilities of landing in each interval:
- $P(S^* = 5) = \frac{5 \cdot P(S=5)}{E[S]} = \frac{5 \cdot (1/3)}{10} = \frac{5}{30} = 1/6$
- $P(S^* = 10) = \frac{10 \cdot P(S=10)}{E[S]} = \frac{10 \cdot (1/3)}{10} = \frac{10}{30} = 1/3$
- $P(S^* = 15) = \frac{15 \cdot P(S=15)}{E[S]} = \frac{15 \cdot (1/3)}{10} = \frac{15}{30} = 1/2$

Then, $E[S^*] = (5 \cdot 1/6) + (10 \cdot 1/3) + (15 \cdot 1/2) = 5/6 + 10/3 + 15/2 = (5 + 20 + 45)/6 = 70/6 = 35/3 \approx 11.67$ minutes. This matches the previous result.

#### Calculation of $E[S^*]$ for Given PDFs

The general formula for the expected length of the interarrival interval containing a random observation, $E[S^*]$, in a renewal process with interarrival times $S$ is:

$E[S^*] = \frac{E[S^2]}{E[S]}$

Where $E[S]$ is the mean of the interarrival times and $E[S^2]$ is the second moment of the interarrival times.

##### 1. For $f_X(x) = \mu e^{-\mu x}$, $x \ge 0$ (Exponential Distribution)

This is the PDF of an Exponential distribution with rate parameter $\mu$. For an Exponential($\mu$) distribution: Mean ($E[S]$): $E[S] = \frac{1}{\mu}$. Variance ($Var(S)$): $Var(S) = \frac{1}{\mu^2}$. Second Moment ($E[S^2]$): $E[S^2] = \frac{1}{\mu^2} + \left(\frac{1}{\mu}\right)^2 = \frac{1}{\mu^2} + \frac{1}{\mu^2} = \frac{2}{\mu^2}$
- $E[S^*] = \frac{E[S^2]}{E[S]} = \frac{\frac{2}{\mu^2}}{\frac{1}{\mu}} = \frac{2}{\mu^2} \cdot \frac{\mu}{1} = \frac{2}{\mu}$

For Exponential interarrival times with rate $\mu$, the expected length of the interarrival interval containing a random observation is $\frac{2}{\mu}$. This is consistent with the result for Poisson processes, as Exponential interarrival times characterize a Poisson process.

##### 2. For $f_X(x) = \frac{\mu^4 x^3 e^{-\mu x}}{3!}$, $x \ge 0$ (Erlang Distribution)

This is the PDF of an Erlang distribution with shape parameter $k=4$ and rate parameter $\mu$.

For an Erlang($k, \mu$) distribution: Mean ($E[S]$): $E[S] = \frac{k}{\mu}$. Variance ($Var(S)$): $Var(S) = \frac{k}{\mu^2}$

In this case, $k=4$: Mean ($E[S]$): $E[S] = \frac{4}{\mu}$. Variance ($Var(S)$): $Var(S) = \frac{4}{\mu^2}$. Second Moment ($E[S^2]$): $E[S^2] = \frac{4}{\mu^2} + \left(\frac{4}{\mu}\right)^2 = \frac{4}{\mu^2} + \frac{16}{\mu^2} = \frac{20}{\mu^2}$
- $E[S^*]$: $E[S^*] = \frac{E[S^2]}{E[S]} = \frac{\frac{20}{\mu^2}}{\frac{4}{\mu}} = \frac{20}{\mu^2} \cdot \frac{\mu}{4} = \frac{5}{\mu}$

For Erlang(4, $\mu$) interarrival times, the expected length of the interarrival interval containing a random observation is $\frac{5}{\mu}$.

Variance ($Var(S)$): $Var(S) = \frac{1}{\mu^2}$. Second Moment ($E[S^2]$): $E[S^2] = \frac{1}{\mu^2} + \left(\frac{1}{\mu}\right)^2 = \frac{1}{\mu^2} + \frac{1}{\mu^2} = \frac{2}{\mu^2}$
- $E[S^*] = \frac{E[S^2]}{E[S]} = \frac{\frac{2}{\mu^2}}{\frac{1}{\mu}} = \frac{2}{\mu^2} \cdot \frac{\mu}{1} = \frac{2}{\mu}$

For Exponential interarrival times with rate $\mu$, the expected length of the interarrival interval containing a random observation is $\frac{2}{\mu}$. This is consistent with the result for Poisson processes, as Exponential interarrival times characterize a Poisson process.

##### 2. For $f_X(x) = \frac{\mu^4 x^3 e^{-\mu x}}{3!}$, $x \ge 0$ (Erlang Distribution)

This is the PDF of an Erlang distribution with shape parameter $k=4$ and rate parameter $\mu$.

For an Erlang($k, \mu$) distribution: Mean ($E[S]$): $E[S] = \frac{k}{\mu}$. Variance ($Var(S)$): $Var(S) = \frac{k}{\mu^2}$

In this case, $k=4$: Mean ($E[S]$): $E[S] = \frac{4}{\mu}$. Variance ($Var(S)$): $Var(S) = \frac{4}{\mu^2}$. Second Moment ($E[S^2]$): $E[S^2] = \frac{4}{\mu^2} + \left(\frac{4}{\mu}\right)^2 = \frac{4}{\mu^2} + \frac{16}{\mu^2} = \frac{20}{\mu^2}$
- $E[S^*]$: $E[S^*] = \frac{E[S^2]}{E[S]} = \frac{\frac{20}{\mu^2}}{\frac{4}{\mu}} = \frac{20}{\mu^2} \cdot \frac{\mu}{4} = \frac{5}{\mu}$

For Erlang(4, $\mu$) interarrival times, the expected length of the interarrival interval containing a random observation is $\frac{5}{\mu}$.

### Sampling Methods

Sampling is the process of selecting a subset of individuals from a larger population. The goal is to obtain a sample that is representative of the population, allowing for generalizations and inferences to be made about the entire group without having to examine every member.

Sampling is essential for several reasons:
*   **Feasibility**: Studying an entire population can be impractical, costly, or time-consuming.
*   **Efficiency**: A well-chosen sample can provide accurate information more quickly and with fewer resources.
*   **Accuracy**: Sometimes, destructive testing is required, making sampling the only viable option.

There are two main categories of sampling methods:

1.  **Probability Sampling**: Every element in the population has a known, non-zero probability of being selected. This allows for statistical inferences to be made about the population and helps in estimating sampling error.
    *   **Simple Random Sampling (SRS)**: Every individual has an equal chance of being selected. This is often done by randomly assigning numbers to each member and then selecting a subset of those numbers.
    *   **Systematic Sampling**: Individuals are selected at regular intervals from a sampling frame (e.g., every 10th person on a list).
    *   **Stratified Sampling**: The population is divided into homogeneous subgroups (strata) based on shared characteristics, and then simple random sampling is performed within each stratum.
    *   **Cluster Sampling**: The population is divided into clusters (usually geographically), and then a random sample of clusters is selected, with all individuals within the chosen clusters being sampled.

2.  **Non-Probability Sampling**: The selection of elements is not based on random chance. This makes it difficult to estimate sampling error and generalize findings to the broader population. These methods are often used in qualitative research, pilot studies, or when probability sampling is not feasible.
    *   **Convenience Sampling**: Individuals are selected based on their easy accessibility and proximity to the researcher.
    *   **Quota Sampling**: Similar to stratified sampling, but selection within subgroups is non-random. Researchers aim to fill a quota for each subgroup.
    *   **Purposive/Judgmental Sampling**: Researchers handpick individuals based on their expertise or specific characteristics relevant to the research.
    *   **Snowball Sampling**: Participants recruit other participants from their network, often used when populations are hard to reach.

Understanding the appropriate sampling method is critical for ensuring the validity and reliability of research findings.

### Poisson Sampling

**Poisson sampling** is a method of sampling where each unit in the population is selected into the sample independently of the other units, based on a pre-assigned probability of inclusion. This method is particularly useful when the size of the final sample is not fixed in advance, but rather follows a random distribution (specifically, a Poisson distribution if the individual probabilities are small and the population is large).

#### How it Works:
- **Assign Inclusion Probabilities:** For each unit $i$ in the population $U = \{U_1, U_2, \dots, U_N\}$, a probability of inclusion $\pi_i$ is assigned. This $\pi_i$ is the probability that unit $U_i$ will be included in the sample.
- **Independent Selection:** Each unit $U_i$ is then independently subjected to a Bernoulli trial: a random number is drawn, and if it is less than or equal to $\pi_i$, the unit is included in the sample; otherwise, it is not.

#### Key Characteristics:
- **Unfixed Sample Size:** Unlike Simple Random Sampling or Stratified Sampling, the sample size in Poisson sampling is a random variable. The expected sample size is $E[n] = \sum_{i=1}^N \pi_i$.
- **Independent Selection:** The inclusion of one unit does not affect the inclusion of any other unit. This simplifies variance estimation.
- **Unequal Probabilities:** It naturally allows for unequal probabilities of selection, which is advantageous when certain units are more important to sample than others (e.g., larger businesses, specific demographics).
- **Approximation by Poisson Distribution:** If the population size $N$ is large and the individual inclusion probabilities $\pi_i$ are small, the actual number of selected units $n$ will approximately follow a Poisson distribution with parameter $\lambda = \sum_{i=1}^N \pi_i$. This is where the name 'Poisson sampling' originates.

#### Advantages:

*   **Flexibility:** Allows for easy adjustment of inclusion probabilities for different units.
*   **Simplicity of Implementation:** Once probabilities are assigned, the selection process for each unit is simple and independent.
*   **Ease of Variance Estimation:** The independence of selections simplifies the calculation of variances of estimators.

#### Disadvantages:

*   **Random Sample Size:** The sample size is not fixed, which can be an issue if a specific budget or precision based on sample size is required. Often, this is managed by repeating the sampling process until a desired sample size is approximately met, or by using techniques like sub-sampling from the initial Poisson sample.
*   **Zero Inclusion Probabilities:** If any $\pi_i = 0$, that unit can never be selected. If $\pi_i = 1$, that unit is always included with certainty.

#### Applications:

Poisson sampling is frequently used in areas such as:

*   **Auditing:** When auditing financial records, auditors might want to sample transactions with higher monetary values (assigning them higher $\pi_i$).
*   **Quality Control:** In manufacturing, items from certain batches or with known potential issues might be sampled with higher probability.
*   **Continuous Sampling:** In situations where the population units become available sequentially, and a decision to include or exclude a unit must be made immediately without knowing future units.

**Example Scenario:** Imagine you have a list of $N$ companies, and you want to sample them for an audit. You decide that larger companies should have a higher chance of being selected. You assign a probability $\pi_i$ to each company $i$ (e.g., proportional to their revenue). Then, for each company, you flip a biased coin with probability $\pi_i$ of heads. If it's heads, the company is in your sample.

#### Plot: Poisson Sampling vs. Simple Random Sampling (SRS)

This visualization compares Poisson sampling with Simple Random Sampling (SRS). We'll observe a population and see which units are selected by each method.

*   **Population**: A set of $N$ items (e.g., integers from 1 to $N$).
*   **Poisson Sampling**: Each item is independently selected with a probability $\pi_i$. The sample size is not fixed.
*   **Simple Random Sampling (SRS)**: A fixed number of $k$ items are selected from the $N$ items, where each combination of $k$ items has an equal chance of being selected. The sample size is fixed.

````python
def simulate_and_plot_sampling_methods(N=100, pi_uniform=0.3, srs_sample_size=None):
    # 1. Define the population
    population = np.arange(1, N + 1)

    # 2. Simulate Poisson Sampling
    # Each unit is selected independently with probability pi_uniform
    poisson_sample_indicators = (np.random.rand(N) < pi_uniform).astype(int)
    poisson_sample = population[poisson_sample_indicators == 1]
    poisson_sample_size = len(poisson_sample)

    # 3. Simulate Simple Random Sampling (SRS)
    # If srs_sample_size is not provided, use the expected sample size from Poisson sampling
    if srs_sample_size is None:
        # For better comparison, let's target the expected Poisson sample size
        # or a reasonable default if N * pi_uniform is too small/large
        srs_sample_size = int(N * pi_uniform) # Expected value
        if srs_sample_size == 0 and N > 0: srs_sample_size = 1 # ensure at least one for small pi
        elif srs_sample_size > N: srs_sample_size = N

    # Ensure srs_sample_size is within bounds
    srs_sample_size = min(srs_sample_size, N)
    srs_sample_size = max(srs_sample_size, 0)

    srs_sample = np.random.choice(population, size=srs_sample_size, replace=False)

# Run the simulation and plot
simulate_and_plot_sampling_methods(N=100, pi_uniform=0.3, srs_sample_size=30)
````
